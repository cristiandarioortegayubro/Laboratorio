## 1️⃣ Introducción

### Neurona Artificial

```
Inputs (x1, x2, ..., xn)
    ↓
Weights (w1, w2, ..., wn)
    ↓
Sum: z = w1*x1 + w2*x2 + ... + b
    ↓
Activation: a = f(z)
    ↓
Output
```

### Funciones de Activación

* **ReLU**: `f(x) = max(0, x)` ← Default para capas ocultas
* **Sigmoid**: `f(x) = 1/(1+e^-x)` ← Clasificación binaria
* **Softmax**: Clasificación multi-clase
* **Tanh**: `f(x) = (e^x - e^-x)/(e^x + e^-x)`

## 2️⃣ Red Neuronal Secuencial

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Definir arquitectura
model = keras.Sequential([
    layers.Dense(128, activation='relu', input_shape=(10,)),  # Capa oculta 1
    layers.Dropout(0.3),                                     # Dropout (previene overfitting)
    layers.Dense(64, activation='relu'),                      # Capa oculta 2
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')                    # Capa de salida (binaria)
])

# Compilar
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Ver arquitectura
model.summary()
```
