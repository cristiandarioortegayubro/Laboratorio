## 🗺️ Análisis Geoespacial en Ciencia de Datos

El **análisis geoespacial** es fundamental en ciencia de datos moderna porque:
- La mayoría de los datos tienen una **componente espacial** (ubicación)
- Permite descubrir **patrones geográficos** ocultos
- Esencial para **optimización de recursos** (rutas, cobertura, logística)
- Mejora la **segmentación** y personalización de clientes
- Clave para **decisiones de negocio** (ubicación de sucursales, expansión)

   - Optimización de rutas

   - Segmentación geográfica de clientes

### 🔷 Sistema de Indexación H3

**H3** es un sistema de cuadrícula geoespacial desarrollado por **Uber** que divide la Tierra en hexágonos jerárquicos.

#### ¿Por qué Hexágonos?

Los hexágonos son superiores a cuadrados o triángulos porque:
- ✅ **Equidistancia**: Todos los vecinos están a la misma distancia
- ✅ **Sin sesgo direccional**: No favorecen ninguna orientación
- ✅ **Mejor aproximación circular**: Ideal para áreas de cobertura
- ✅ **Agregación eficiente**: Fácil agrupar zonas

#### Resoluciones H3

H3 usa 16 niveles de resolución (0-15):

| Resolución | Área por Hexágono | Uso Típico |
|------------|-------------------|------------|
| 0 | ~4,000 km² | Continentes |
| 3 | ~100 km² | Países |
| 5 | ~25 km² | Ciudades grandes |
| 7 | ~5 km² | Barrios |
| **9** | **~0.1 km²** | **Zonas comerciales** ⭐ |
| 11 | ~0.02 km² | Manzanas |
| 13 | ~0.001 km² | Edificios |
| 15 | ~0.00001 km² | Habitaciones |

**Para análisis de retail y negocios locales, la resolución 9 es ideal.**

#### Ventajas de H3

1. **Agregación rápida**: `GROUP BY h3_index`
2. **Cálculo de distancias**: `h3.grid_distance(h3_a, h3_b)`
3. **Vecindario**: `h3.grid_disk(h3_index, radio)`
4. **Jerárquico**: Fácil cambiar de resolución
5. **Estándar abierto**: Usado por Uber, DoorDash, Amazon

### 💡 Ejemplo: Análisis de Cobertura de Panadería

**Problema**: ¿Cuántos clientes están a menos de 500m de cada sucursal?

**Solución con H3**:

```python
import h3
import pandas as pd

# 1. Convertir coordenadas GPS a índices H3
df_sucursales['h3'] = df_sucursales.apply(
    lambda row: h3.latlng_to_cell(row['lat'], row['lon'], 9),
    axis=1
)

df_clientes['h3'] = df_clientes.apply(
    lambda row: h3.latlng_to_cell(row['lat'], row['lon'], 9),
    axis=1
)

# 2. Para cada sucursal, obtener hexágonos vecinos (radio ~5 = 500m)
for _, sucursal in df_sucursales.iterrows():
    area_cobertura = h3.grid_disk(sucursal['h3'], 5)
    
    # 3. Contar clientes en esa área
    clientes_cercanos = df_clientes[
        df_clientes['h3'].isin(area_cobertura)
    ]
    
    print(f"{sucursal['nombre']}: {len(clientes_cercanos)} clientes")
```

**Resultado**:
- Sucursal Centro: 142 clientes
- Sucursal Este: 98 clientes  
- Sucursal Sur: 76 clientes

✅ **Insight**: La sucursal Centro tiene 45% más cobertura → priorizar inversión

## 1️⃣ Introducción al Análisis de Datos

### ¿Qué es el Análisis de Datos?

El **análisis de datos** es el proceso de inspeccionar, limpiar, transformar y modelar datos con el objetivo de descubrir información útil, llegar a conclusiones y apoyar la toma de decisiones.

### Tipos de Análisis de Datos

#### 📈 **Análisis Descriptivo** (¿Qué pasó?)
* Resume datos históricos
* Proporciona contexto sobre el pasado
* Ejemplos: reportes de ventas, dashboards de KPIs
* **Herramientas**: agregaciones, estadísticas descriptivas, visualizaciones

#### 🔍 **Análisis Diagnóstico** (¿Por qué pasó?)
* Identifica causas y relaciones
* Profundiza en patrones y anomalías
* Ejemplos: análisis de caída en ventas, identificación de outliers
* **Herramientas**: correlaciones, drill-down, comparaciones

#### 🔮 **Análisis Predictivo** (¿Qué pasará?)
* Utiliza datos históricos para predecir futuros
* Aplica modelos estadísticos y machine learning
* Ejemplos: pronóstico de demanda, predicción de churn
* **Herramientas**: regresión, series temporales, ML

#### 💡 **Análisis Prescriptivo** (¿Qué debemos hacer?)
* Recomienda acciones basadas en datos
* Optimiza decisiones de negocio
* Ejemplos: optimización de inventario, pricing dinámico
* **Herramientas**: optimización, simulación, IA

## 2️⃣ Computación en la Nube para Data Science

### ¿Por qué Cloud Computing?

**Ventajas clave:**

1. **🚀 Escalabilidad**: Ajusta recursos según demanda (datos grandes o pequeños)
2. **💰 Costo-efectividad**: Paga solo por lo que usas
3. **⚡ Velocidad**: Implementación rápida sin esperar hardware
4. **🌍 Accesibilidad**: Trabaja desde cualquier lugar
5. **🔄 Colaboración**: Equipos distribuidos trabajando juntos

### Modelos de Servicio Cloud

* **IaaS** (Infrastructure as a Service): Control total del entorno
* **PaaS** (Platform as a Service): Plataforma lista para desarrollo
* **SaaS** (Software as a Service): Software listo para usar

**Databricks es una plataforma PaaS** diseñada específicamente para análisis de datos y machine learning.

## 3️⃣ Databricks como Plataforma Unificada

### ¿Qué es Databricks?

**Databricks** es una plataforma unificada para:
* Ingeniería de datos
* Análisis de datos
* Machine Learning
* Business Intelligence

### Arquitectura de Databricks

```
┌─────────────────────────────────────────┐
│     Databricks Workspace (UI/IDE)       │
│  Notebooks | SQL Editor | Dashboards    │
└─────────────────────────────────────────┘
                   ↓
┌─────────────────────────────────────────┐
│        Databricks Runtime (DBR)         │
│  Apache Spark + Optimizaciones          │
└─────────────────────────────────────────┘
                   ↓
┌─────────────────────────────────────────┐
│         Delta Lake (Storage)            │
│  ACID Transactions + Time Travel        │
└─────────────────────────────────────────┘
```

### Componentes Clave

1. **Workspace**: Entorno colaborativo (notebooks, código, visualizaciones)
2. **Clusters**: Recursos computacionales (escalables, configurables)
3. **Delta Lake**: Capa de almacenamiento con ACID transactions
4. **MLflow**: Gestión del ciclo de vida de ML
5. **Unity Catalog**: Gobernanza y seguridad de datos

## 4️⃣ Herramientas de Análisis: Pandas y PySpark

### Pandas: Análisis de Datos en Memoria

**Pandas** es ideal para:
* Datasets pequeños/medianos (hasta ~1GB)
* Prototipado rápido
* Análisis exploratorio

**Ventajas:**
* API intuitiva y fácil de aprender
* Gran ecosistema de librerías compatibles
* Documentación excelente

**Limitaciones:**
* Procesamiento en una sola máquina
* Limitado por memoria RAM
* No escala a Big Data

### PySpark: Análisis Distribuido

**PySpark** (Python + Apache Spark) es ideal para:
* Datasets grandes (GB, TB, PB)
* Procesamiento distribuido
* Pipelines de producción

**Ventajas:**
* Escalabilidad horizontal (múltiples nodos)
* Procesamiento en memoria distribuida
* Soporte para streaming y batch

**Trade-offs:**
* Curva de aprendizaje más pronunciada
* Overhead en datasets pequeños

### ¿Cuándo usar cada una?

| Criterio | Pandas | PySpark |
|----------|--------|----------|
| Tamaño datos | < 1 GB | > 1 GB |
| Velocidad desarrollo | ⚡⚡⚡ | ⚡⚡ |
| Escalabilidad | ❌ | ✅ |
| Producción | Para datasets pequeños | Recomendado |
| Colaboración | Local | Distribuido |

## 5️⃣ SQL en el Análisis de Datos

### ¿Por qué SQL?

**SQL** (Structured Query Language) es fundamental porque:

1. **📊 Estándar universal**: Lenguaje común en todas las organizaciones
2. **💪 Potente para análisis**: Agregaciones, joins, window functions
3. **🤝 Accesible**: Entendible por equipos técnicos y de negocio
4. **⚡ Optimizado**: Motores SQL son altamente eficientes

### SQL en Databricks

Databricks ofrece **Databricks SQL**, optimizado para:
* Consultas sobre Delta Lake
* Visualizaciones y dashboards
* Integración con BI tools

**Puedes usar SQL en notebooks:**
```sql
%sql
SELECT categoria, COUNT(*) as total
FROM ventas
GROUP BY categoria
ORDER BY total DESC
```

### SQL vs. Python/PySpark

**No son excluyentes** - úsalos complementariamente:

* **SQL**: Consultas, agregaciones, exploraciones rápidas
* **Python/PySpark**: Transformaciones complejas, ML, automatización

**Mejor práctica**: Comienza con SQL para explorar, luego Python para transformaciones avanzadas.

## 6️⃣ Ciclo de Vida de un Proyecto de Datos

### Fases del Proyecto

```
1. ENTENDIMIENTO DEL NEGOCIO
   ↓
2. ADQUISICIÓN Y EXPLORACIÓN DE DATOS
   ↓
3. PREPARACIÓN Y LIMPIEZA
   ↓
4. ANÁLISIS Y MODELADO
   ↓
5. VISUALIZACIÓN E INSIGHTS
   ↓
6. IMPLEMENTACIÓN Y MONITOREO
```

### 1. Entendimiento del Negocio
* Define el problema a resolver
* Establece métricas de éxito
* Identifica stakeholders
* Comprende el contexto

### 2. Adquisición y Exploración
* Identifica fuentes de datos
* Carga inicial (ETL/ELT)

* Documentación inicial

### 3. Preparación y Limpieza

* Detección de outliers
* Normalización y transformaciones
* Feature engineering básico

### 4. Análisis y Modelado
* Análisis estadístico
* Correlaciones y patrones
* Modelos predictivos (si aplica)
* Validación de hipótesis

### 5. Visualización e Insights
* Dashboards interactivos
* Reportes ejecutivos
* Storytelling con datos
* Recomendaciones accionables

### 6. Implementación y Monitoreo
* Deploy a producción
* Automatización de pipelines
* Monitoreo continuo
* Iteración y mejora

## 7️⃣ Mejores Prácticas en Análisis de Datos

### 📝 Documentación
* **Documenta tu código**: Comentarios claros y concisos
* **Notebooks narrativos**: Explica el "por qué" no solo el "cómo"
* **README files**: Contexto del proyecto y estructura
* **Diccionario de datos**: Define cada variable

### ⚡ Performance
* **Muestra antes de todo**: Trabaja con samples en exploración
* **Cache inteligente**: Usa `.cache()` en DataFrames reutilizados
* **Particionamiento**: Organiza datos por claves frecuentes
* **Evita shuffles innecesarios**: Minimiza operaciones costosas

### 🔄 Reproducibilidad
* **Control de versiones**: Git para código y notebooks
* **Semillas aleatorias**: Fija seeds en muestreos y ML
* **Dependencias claras**: Especifica versiones de librerías
* **Datos versionados**: Delta Lake time travel para auditoría

### ✅ Validación de Datos
* **Schema enforcement**: Valida tipos de datos
* **Data quality checks**: Nulos, duplicados, rangos
* **Unit tests**: Prueba transformaciones críticas
* **Assertions**: Verifica supuestos del negocio

### 🔒 Seguridad y Gobernanza
* **Control de acceso**: Permisos por rol
* **Datos sensibles**: Encriptación y enmascaramiento
* **Auditoría**: Registra accesos y modificaciones
* **Compliance**: Cumple regulaciones (GDPR, etc.)

### 🧪 Experimentación
* **Workspace organizado**: Carpetas por proyecto
* **Notebooks modulares**: Funciones reutilizables
* **Parámetros externos**: Configuración en variables
* **Iteración rápida**: Prototipa antes de optimizar
