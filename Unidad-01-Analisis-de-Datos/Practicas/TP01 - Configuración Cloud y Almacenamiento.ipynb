## Parte 3: Carga de Datasets CSV con Pandas

4. **ventas.csv**: Transacciones de ventas

```python
# Cargar el dataset de clientes
df_clientes = pd.read_csv(ruta_datos + 'clientes.csv')

print("✅ Dataset clientes.csv cargado exitosamente")
print(f"\n📊 Dimensiones: {df_clientes.shape[0]} filas x {df_clientes.shape[1]} columnas")
print(f"\n📌 Columnas: {list(df_clientes.columns)}")
print(f"\n📄 Primeras 10 filas:")
df_clientes.head(10)
```

```python
# Cargar el dataset de ventas
df_ventas = pd.read_csv(ruta_datos + 'ventas.csv')

print("✅ Dataset ventas.csv cargado exitosamente")
print(f"\n📊 Dimensiones: {df_ventas.shape[0]} filas x {df_ventas.shape[1]} columnas")
print(f"\n📌 Columnas: {list(df_ventas.columns)}")
print(f"\n📄 Primeras 5 filas:")
df_ventas.head()
```

```python
# Cargar el dataset de detalles de ventas
df_detalles_ventas = pd.read_csv(ruta_datos + 'detalles_ventas.csv')

print("✅ Dataset detalles_ventas.csv cargado exitosamente")
print(f"\n📊 Dimensiones: {df_detalles_ventas.shape[0]} filas x {df_detalles_ventas.shape[1]} columnas")
print(f"\n📌 Columnas: {list(df_detalles_ventas.columns)}")
print(f"\n📄 Primeras 5 filas:")
df_detalles_ventas.head()
```

## Parte 4: Inspección Básica de Datos

### 🔍 Métodos esenciales de Pandas

Ahora que tenemos los datos cargados, vamos a inspeccionarlos usando los métodos más importantes de Pandas:

* `.info()`: Información general del DataFrame
* `.describe()`: Estadísticas descriptivas
* `.dtypes`: Tipos de datos de cada columna
* `.isnull().sum()`: Valores faltantes
* `.value_counts()`: Conteo de valores únicos

```python
# Información general del dataset de productos
print("📊 INFORMACIÓN GENERAL - PRODUCTOS")
print("=" * 60)
df_productos.info()

print("\n📈 ESTADÍSTICAS DESCRIPTIVAS")
print("=" * 60)
display(df_productos.describe())

print("\n📊 CONTEO POR CATEGORÍA")
print("=" * 60)
print(df_productos['categoria'].value_counts())
```

```python
# Verificar valores faltantes en todos los datasets
print("🔍 VALORES FALTANTES POR DATASET")
print("=" * 60)

print("\n1. Productos:")
print(df_productos.isnull().sum())

print("\n2. Sucursales:")
print(df_sucursales.isnull().sum())

print("\n3. Clientes:")
print(df_clientes.isnull().sum())

print("\n4. Ventas:")
print(df_ventas.isnull().sum())

print("\n5. Detalles de Ventas:")
print(df_detalles_ventas.isnull().sum())
```

```python
# Análisis rápido de ventas
print("📊 ANÁLISIS RÁPIDO DE VENTAS")
print("=" * 60)

print(f"\nTotal de ventas: {len(df_ventas):,}")
print(f"Total de líneas de venta: {len(df_detalles_ventas):,}")
print(f"Promedio de productos por venta: {len(df_detalles_ventas) / len(df_ventas):.2f}")

print("\n💰 Facturación:")
print(f"Total: ${df_ventas['total'].sum():,.2f}")
print(f"Promedio por venta: ${df_ventas['total'].mean():,.2f}")
print(f"Venta mínima: ${df_ventas['total'].min():,.2f}")
print(f"Venta máxima: ${df_ventas['total'].max():,.2f}")

print("\n🏢 Ventas por sucursal:")
print(df_ventas.groupby('sucursal_id')['total'].agg(['count', 'sum', 'mean']).round(2))
```

## Parte 5: Ejercicios Prácticos

### ✍️ Ejercicios para Resolver

Ahora es tu turno. Completa los siguientes ejercicios utilizando los datasets cargados.

#### **Ejercicio 1**: Productos más caros
Encuentra los 5 productos más caros del catálogo y muéstralos con su nombre, categoría y precio.

```python
# EJERCICIO 1: Productos más caros
# Pista: usa .nlargest() o .sort_values()

# Tu código aquí:
top_5_caros = df_productos.nlargest(5, 'precio_unitario')[['nombre', 'categoria', 'precio_unitario']]

print("💰 TOP 5 PRODUCTOS MÁS CAROS")
print("=" * 60)
display(top_5_caros)
```

#### **Ejercicio 2**: Clientes VIP
Cuántos clientes son VIP y cuál es su porcentaje sobre el total de clientes?

```python
# EJERCICIO 2: Clientes VIP
# Pista: usa .value_counts() o .sum()

# Tu código aquí:
total_clientes = len(df_clientes)
clientes_vip = df_clientes['es_vip'].sum()
porcentaje_vip = (clientes_vip / total_clientes) * 100

print("🌟 CLIENTES VIP")
print("=" * 60)
print(f"Total clientes: {total_clientes}")
print(f"Clientes VIP: {clientes_vip}")
print(f"Porcentaje VIP: {porcentaje_vip:.2f}%")
```

#### **Ejercicio 3**: Ventas por mes
Calcula el total de ventas (cantidad de transacciones) por mes del año 2024.

```python
# EJERCICIO 3: Ventas por mes en 2024
# Pista: filtra por año y agrupa por mes

# Tu código aquí:
ventas_2024 = df_ventas[df_ventas['anio'] == 2024]
ventas_por_mes = ventas_2024.groupby('mes').size().reset_index(name='cantidad_ventas')

print("📅 VENTAS POR MES - AÑO 2024")
print("=" * 60)
display(ventas_por_mes)
```
