## Ejercicio: Crear Presentación Ejecutiva

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd

# Simular datos de ventas
df = pd.DataFrame({
    'mes': pd.date_range('2024-01', periods=12, freq='M'),
    'ventas': [100, 110, 115, 105, 120, 130, 125, 140, 145, 150, 155, 160],
    'objetivo': [110] * 12
})

# Slide 1: KPI Principal
fig1 = go.Figure()
fig1.add_trace(go.Indicator(
    mode='number+delta',
    value=df['ventas'].iloc[-1],
    delta={'reference': df['ventas'].iloc[-2], 'relative': True},
    title={'text': 'Ventas Mes Actual (K$)'},
    number={'prefix': ', 'suffix': 'K'}
))
fig1.update_layout(title='Resumen Ejecutivo', height=300)
fig1.show()

# Slide 2: Tendencia
fig2 = go.Figure()
fig2.add_trace(go.Scatter(x=df['mes'], y=df['ventas'], mode='lines+markers', name='Real'))
fig2.add_trace(go.Scatter(x=df['mes'], y=df['objetivo'], mode='lines', name='Objetivo', line=dict(dash='dash')))
fig2.add_annotation(x=df['mes'].iloc[6], y=df['ventas'].iloc[6], text='Lanzamiento', showarrow=True)
fig2.update_layout(title='Evolución de Ventas 2024', xaxis_title='Mes', yaxis_title='Ventas (K$)')
fig2.show()

# Slide 3: Segmentación
categorias = pd.DataFrame({
    'categoria': ['A', 'B', 'C', 'D'],
    'ventas': [50, 35, 25, 10],
    'crecimiento': [15, 8, -5, 2]
})

fig3 = go.Figure()
fig3.add_trace(go.Bar(x=categorias['categoria'], y=categorias['ventas'], marker_color=['green', 'blue', 'orange', 'red']))
fig3.update_layout(title='Ventas por Categoría', xaxis_title='Categoría', yaxis_title='Ventas (K$)')
fig3.show()

print("✅ Presentación ejecutiva generada")
print("💡 Insights: Ventas crecieron 3% vs mes anterior, Categoría A lidera")
```
