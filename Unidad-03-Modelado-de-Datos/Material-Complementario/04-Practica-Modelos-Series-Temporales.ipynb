# 💻 Módulo 04 - Práctica: Series Temporales

## Ejercicios con ARIMA, Prophet y LSTM

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from prophet import Prophet
import warnings
warnings.filterwarnings('ignore')

# Generar serie temporal sintética
np.random.seed(42)
dates = pd.date_range('2020-01-01', periods=365*3, freq='D')
trend = np.linspace(100, 200, len(dates))
seasonality = 10 * np.sin(2 * np.pi * np.arange(len(dates)) / 365)
noise = np.random.randn(len(dates)) * 5
ts = pd.Series(trend + seasonality + noise, index=dates)

print("✅ Serie temporal generada: 3 años de datos diarios")

# Split train/test
train_size = int(len(ts) * 0.8)
train = ts[:train_size]
test = ts[train_size:]

print(f"Train: {len(train)} días, Test: {len(test)} días")

# Ejercicio 1: ARIMA
model_arima = ARIMA(train, order=(1, 1, 1))
model_fit = model_arima.fit()
forecast_arima = model_fit.forecast(steps=len(test))

mae_arima = np.mean(np.abs(test.values - forecast_arima))
print(f"\n✅ ARIMA MAE: {mae_arima:.2f}")

# Ejercicio 2: Prophet
df_prophet = pd.DataFrame({
    'ds': train.index,
    'y': train.values
})

model_prophet = Prophet(yearly_seasonality=True, daily_seasonality=False)
model_prophet.fit(df_prophet)

future = pd.DataFrame({'ds': test.index})
forecast_prophet = model_prophet.predict(future)
forecast_prophet_values = forecast_prophet['yhat'].values

mae_prophet = np.mean(np.abs(test.values - forecast_prophet_values))
print(f"✅ Prophet MAE: {mae_prophet:.2f}")

# Ejercicio 3: Visualización comparativa
fig, axes = plt.subplots(2, 1, figsize=(14, 8))

# ARIMA
axes[0].plot(train.index, train.values, label='Train', color='blue')
axes[0].plot(test.index, test.values, label='Test (Real)', color='green')
axes[0].plot(test.index, forecast_arima, label='ARIMA Forecast', color='red', linestyle='--')
axes[0].set_title(f'ARIMA - MAE: {mae_arima:.2f}')
axes[0].legend()
axes[0].grid(True)

# Prophet
axes[1].plot(train.index, train.values, label='Train', color='blue')
axes[1].plot(test.index, test.values, label='Test (Real)', color='green')
axes[1].plot(test.index, forecast_prophet_values, label='Prophet Forecast', color='red', linestyle='--')
axes[1].set_title(f'Prophet - MAE: {mae_prophet:.2f}')
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
display(plt.gcf())
plt.close()

# Ejercicio 4: Comparación de métricas
resultados = pd.DataFrame({
    'Modelo': ['ARIMA', 'Prophet'],
    'MAE': [mae_arima, mae_prophet],
    'RMSE': [
        np.sqrt(np.mean((test.values - forecast_arima)**2)),
        np.sqrt(np.mean((test.values - forecast_prophet_values)**2))
    ]
})

print("\nComparación de Modelos:")
print(resultados)

print("\n✅ Ejercicios completados")
print("💡 Prophet captura mejor la estacionalidad anual")
print("💡 ARIMA es más simple y rápido")
```
