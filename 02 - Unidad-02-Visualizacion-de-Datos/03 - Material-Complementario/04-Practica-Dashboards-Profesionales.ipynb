## Dashboard Interactivo Completo

```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy as np

# Generar datos de ejemplo
np.random.seed(42)
dates = pd.date_range('2024-01', periods=365, freq='D')
df = pd.DataFrame({
    'fecha': dates,
    'ventas': np.random.normal(1000, 200, 365).cumsum(),
    'categoria': np.random.choice(['A', 'B', 'C'], 365)
})

# Crear app
app = dash.Dash(__name__)

# Layout
app.layout = html.Div([
    html.H1('Dashboard de Ventas 2024', style={'textAlign': 'center'}),
    
    # KPIs Row
    html.Div([
        html.Div([
            html.H3(f"${df['ventas'].sum():,.0f}"),
            html.P('Ventas Totales')
        ], style={'textAlign': 'center', 'padding': '20px', 'backgroundColor': '#e3f2fd'}),
        
        html.Div([
            html.H3(f"${df['ventas'].mean():,.0f}"),
            html.P('Promedio Diario')
        ], style={'textAlign': 'center', 'padding': '20px', 'backgroundColor': '#f3e5f5'}),
        
        html.Div([
            html.H3(f"+{((df['ventas'].iloc[-1] / df['ventas'].iloc[0] - 1) * 100):.1f}%"),
            html.P('Crecimiento')
        ], style={'textAlign': 'center', 'padding': '20px', 'backgroundColor': '#e8f5e9'})
    ], style={'display': 'flex', 'justifyContent': 'space-around', 'marginBottom': '30px'}),
    
    # Filtro
    html.Div([
        html.Label('Filtrar por Categoría:'),
        dcc.Dropdown(
            id='filtro-categoria',
            options=[{'label': 'Todas', 'value': 'ALL'}] + 
                    [{'label': cat, 'value': cat} for cat in df['categoria'].unique()],
            value='ALL'
        )
    ], style={'width': '300px', 'marginBottom': '20px'}),
    
    # Gráficos
    dcc.Graph(id='grafico-tendencia'),
    
    html.Div([
        dcc.Graph(id='grafico-categoria', style={'width': '48%', 'display': 'inline-block'}),
        dcc.Graph(id='grafico-distribucion', style={'width': '48%', 'display': 'inline-block'})
    ])
])

# Callbacks
@app.callback(
    [Output('grafico-tendencia', 'figure'),
     Output('grafico-categoria', 'figure'),
     Output('grafico-distribucion', 'figure')],
    Input('filtro-categoria', 'value')
)
def update_graphs(categoria):
    df_filtrado = df if categoria == 'ALL' else df[df['categoria'] == categoria]
    
    # Tendencia
    fig1 = px.line(df_filtrado, x='fecha', y='ventas', title='Evolución de Ventas')
    fig1.update_layout(template='plotly_white')
    
    # Por categoría
    cat_agg = df.groupby('categoria')['ventas'].sum().reset_index()
    fig2 = px.bar(cat_agg, x='categoria', y='ventas', title='Ventas por Categoría')
    
    # Distribución
    fig3 = px.histogram(df_filtrado, x='ventas', nbins=30, title='Distribución de Ventas Diarias')
    
    return fig1, fig2, fig3

print("✅ Dashboard configurado")
print("💡 Ejecuta: app.run_server(debug=True, port=8050)")
print("🌐 Abre: http://localhost:8050")

# Para correr:
# app.run_server(debug=True, port=8050)
```

## Ejercicio Adicional: Dashboard con Subplots

```python
from plotly.subplots import make_subplots

def crear_dashboard_completo(df):
    fig = make_subplots(
        rows=3, cols=2,
        subplot_titles=('KPI: Ventas Totales', 'Tendencia Temporal',
                       'Top Categorías', 'Distribución',
                       'Heatmap Semanal', 'Forecast'),
        specs=[[{'type': 'indicator'}, {'type': 'scatter'}],
               [{'type': 'bar'}, {'type': 'histogram'}],
               [{'type': 'heatmap'}, {'type': 'scatter'}]]
    )
    
    # KPI
    fig.add_trace(go.Indicator(
        mode='number+delta',
        value=df['ventas'].sum(),
        delta={'reference': df['ventas'].sum() * 0.9}
    ), row=1, col=1)
    
    # Tendencia
    fig.add_trace(go.Scatter(x=df['fecha'], y=df['ventas'], mode='lines'), row=1, col=2)
    
    # Categorías
    cat_sum = df.groupby('categoria')['ventas'].sum()
    fig.add_trace(go.Bar(x=cat_sum.index, y=cat_sum.values), row=2, col=1)
    
    # Distribución
    fig.add_trace(go.Histogram(x=df['ventas']), row=2, col=2)
    
    fig.update_layout(height=1200, showlegend=False, title_text='Dashboard Completo')
    return fig

fig_completo = crear_dashboard_completo(df)
fig_completo.show()

print("✅ Dashboard con subplots generado")
```
