## 1️⃣ Principios de Diseño de Dashboards

### Regla 5-3-1

* **5 segundos**: Usuario entiende propósito  
* **3 métricas**: Máximo por sección  
* **1 acción**: Clara por insight

### Layout

**Patrón F**: Usuarios leen en F
```
KPI1  KPI2  KPI3
╔═══════════════╗
║ Gráfico      ║
║ Principal    ║
╚═══════════════╝
╔═══════╗ ╔═══════╗
║ Seg 1 ║ ║ Seg 2 ║
╚═══════╝ ╚═══════╝
```

### Dashboard Básico

```python
import dash
from dash import dcc, html
import plotly.express as px
import pandas as pd

app = dash.Dash(__name__)

df = pd.read_csv('ventas.csv')

app.layout = html.Div([
    html.H1('Dashboard de Ventas'),
    
    html.Div([
        html.H3(f"${df['monto'].sum():,.0f}"),
        html.P('Ventas Totales')
    ], style={'textAlign': 'center'}),
    
    dcc.Graph(
        figure=px.line(df, x='fecha', y='monto', title='Tendencia')
    ),
    
    dcc.Graph(
        figure=px.bar(df.groupby('categoria')['monto'].sum().reset_index(),
                      x='categoria', y='monto')
    )
])

if __name__ == '__main__':
    app.run_server(debug=True, port=8050)
```

---

## 3️⃣ Interactividad con Callbacks

```python
from dash import Input, Output

@app.callback(
    Output('grafico-ventas', 'figure'),
    Input('filtro-categoria', 'value')
)
def update_graph(categoria_seleccionada):
    df_filtrado = df[df['categoria'] == categoria_seleccionada]
    fig = px.line(df_filtrado, x='fecha', y='monto')
    return fig
```

## 4️⃣ Mejores Prácticas

### Performance

✅ **Hacer**:
* Cachear datos (`@cache.memoize`)
* Agregar antes de visualizar
* Lazy loading de secciones

❌ **Evitar**:
* Queries en cada callback
* Gráficos con > 10K puntos sin agregar
* Actualizaciones demasiado frecuentes

### UX

* Loading spinners
* Mensajes de error claros
* Responsive design
* Tooltips informativos
