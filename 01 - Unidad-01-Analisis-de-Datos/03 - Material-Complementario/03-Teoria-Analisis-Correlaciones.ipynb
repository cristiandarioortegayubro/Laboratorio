## 1️⃣ ¿Qué es la Correlación?

La **correlación** mide la **relación lineal** entre dos variables.

**Valores:**
* **r = +1**: Correlación positiva perfecta (↗️)
* **r = 0**: Sin correlación
* **r = -1**: Correlación negativa perfecta (↘️)

### 📊 Tipos de Correlación

#### **Pearson (paramétrica)**
* Mide relación **lineal**
* Asume **distribución normal**
* Sensible a **outliers**
```python
corr_pearson = df['var1'].corr(df['var2'], method='pearson')
```

#### **Spearman (no paramétrica)**
* Basada en **rankings**
* **Robusta a outliers**
* Detecta relaciones **monotónicas** (no solo lineales)
```python
corr_spearman = df['var1'].corr(df['var2'], method='spearman')
```

#### **Kendall (no paramétrica)**
* Basada en **concordancia de pares**
* Más robusta que Spearman
* Mejor para **datasets pequeños**
```python
corr_kendall = df['var1'].corr(df['var2'], method='kendall')
```

## 2️⃣ Interpretación de Correlaciones

| |r| | Interpretación | Acción en ML |
|---------|----------------|----------------|
| **0.00 - 0.19** | Muy débil | Probablemente no predictivo |
| **0.20 - 0.39** | Débil | Puede ser útil combinado |
| **0.40 - 0.59** | Moderada | Potencialmente predictivo |
| **0.60 - 0.79** | Fuerte | Muy predictivo |
| **0.80 - 1.00** | Muy fuerte | ⚠️ Revisar multicolinealidad |

**⚠️ Advertencias:**
* **Correlación ≠ Causalidad**
* Puede haber **relaciones no lineales** (usar Spearman)
* **Outliers** afectan Pearson

## 3️⃣ Multicolinealidad

**Problema**: Dos o más variables predictoras están **altamente correlacionadas**.

**Impacto en modelos ML:**
* ❌ Coeficientes **inestables** (regresión)
* ❌ **Interpretación confusa** (feature importance)
* ❌ **Varianza inflada** de estimadores

**🔍 Detección:**

## 4️⃣ Selección de Features por Correlación

**Estrategia:**

```python
# 1. Correlación con target
target_corr = df.corr()['target'].abs().sort_values(ascending=False)
print("Top features correlacionados con target:")
print(target_corr.head(10))

# 2. Eliminar features con baja correlación con target
threshold = 0.1
low_corr_features = target_corr[target_corr < threshold].index.tolist()
df_reduced = df.drop(low_corr_features, axis=1)

# 3. Eliminar features altamente correlacionados entre sí
corr_matrix = df_reduced.corr().abs()
upper_triangle = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop = [col for col in upper_triangle.columns if any(upper_triangle[col] > 0.85)]
df_final = df_reduced.drop(to_drop, axis=1)

print(f"\nFeatures originales: {df.shape[1]}")
print(f"Features finales: {df_final.shape[1]}")
```
