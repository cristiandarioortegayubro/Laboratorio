### 🗺️ Módulo 02: Mapas y Visualización Geoespacial

**🎯 Objetivos:**
* Crear mapas interactivos con Plotly
* Aplicar indexación H3 para análisis espacial
* Visualizar densidad y zonas calientes
* Implementar casos de uso geoespaciales

**📓 Notebooks:**

#### [02-Teoria-Mapas-Visualizacion-Geoespacial](#notebook-1402451318137592)
* **Contenido**:
  * **Tipos de Mapas**:
    * Scatter Mapbox (puntos)
    * Density Mapbox (heatmap)
    * Choropleth (áreas coloreadas)
    * Line Mapbox (rutas)
  * **H3: Indexación Geoespacial**
    * Sistema jerárquico de hexágonos (Uber)
    * Resoluciones (0-15) y casos de uso
    * Ventajas: agregación rápida, jerarquía, vecindad
    * Aplicación en DataFrames
  * **Casos de Uso**:
    * Optimización de ubicación de sucursales
    * Análisis de rutas de delivery
    * Segmentación geográfica de clientes
  * Estilos de mapas (OpenStreetMap, Carto, Stamen)
  * Mejores prácticas

#### [02-Practica-Mapas-Visualizacion-Geoespacial](#notebook-1402451318137593)
* **Ejercicios**:
  1. Scatter Mapbox básico con datos de Mendoza
  2. Agregación por zona H3
  3. Heatmap de densidad
  4. Visualización de zonas calientes
* **Datasets**: Simulación geoespacial de ventas (lat/lon + H3)

**⏱️ Duración**: 90 minutos

### 📖 Módulo 03: Storytelling con Datos

**🎯 Objetivos:**
* Estructurar narrativas efectivas con datos
* Elegir visualizaciones apropiadas por objetivo
* Aplicar principios de diseño visual
* Crear presentaciones ejecutivas

**📓 Notebooks:**

#### [03-Teoria-Storytelling-con-Datos](#notebook-1402451318137594)
* **Contenido**:
  * **Estructura de Historia de Datos**:
    * Contexto (¿Por qué es importante?)
    * Complicación (¿Qué está pasando?)
    * Resolución (¿Qué hacer?)
  * **Elegir Visualización Correcta**:
    * Tendencia → Line chart
    * Comparación → Bar chart
    * Distribución → Histogram, Box plot
    * Correlación → Scatter plot
    * Proporción → Pie, Sunburst
    * Geoespacial → Mapa
    * KPI → Indicator, Gauge
  * **Principios de Diseño**:
    * Ley de Gestalt (proximidad, similitud, continuidad)
    * Jerarquía visual (primario, secundario, terciario)
    * Uso efectivo del color
  * **Framework: Presentación Ejecutiva**
    * Slide 1: Resumen ejecutivo (1 KPI grande)
    * Slide 2: Tendencias
    * Slide 3: Segmentación
    * Slide 4: Recomendaciones
  * Checklist de calidad

#### [03-Practica-Storytelling-con-Datos](#notebook-1402451318137595)
* **Ejercicio**: Crear presentación ejecutiva completa
  1. KPI principal con indicator
  2. Tendencia temporal con anotaciones
  3. Segmentación visual por categoría
  4. Interpretación de insights
* **Entregable**: Dashboard narrativo con storytelling

**⏱️ Duración**: 60-75 minutos

### 📊 Módulo 04: Dashboards Profesionales

**🎯 Objetivos:**
* Diseñar dashboards efectivos
* Implementar con Plotly Dash
* Crear interactividad con callbacks
* Optimizar performance y desplegar

**📓 Notebooks:**

#### [04-Teoria-Dashboards-Profesionales](#notebook-1402451318137596)
* **Contenido**:
  * **Principios de Diseño**:
    * Regla 5-3-1 (5 seg comprensión, 3 métricas max, 1 acción)
    * Patrón F de lectura
    * Layout efectivo
  * **Plotly Dash**:
    * Instalación y setup
    * Estructura básica (app.layout)
    * Componentes (html, dcc.Graph, dcc.Dropdown)
  * **Interactividad con Callbacks**:
    * Decorador @app.callback
    * Input/Output linking
    * Filtros dinámicos
  * **Mejores Prácticas**:
    * Performance (caching, agregación, lazy loading)
    * UX (loading spinners, mensajes de error)
    * Responsive design
  * **Databricks Integration**
  * Despliegue (Databricks Apps, Heroku, AWS)
  * Checklist dashboard profesional

#### [04-Practica-Dashboards-Profesionales](#notebook-1402451318137597)
* **Ejercicios**:
  1. Dashboard interactivo completo con Dash
    * KPIs en top (3 métricas)
    * Filtro por categoría (dropdown)
    * Gráfico de tendencia
    * Gráfico de categoría
    * Distribución
  2. Dashboard con subplots
    * 6 gráficos en grid (3x2)
    * Indicator, scatter, bar, histogram, heatmap, forecast
* **Entregable**: Dashboard funcional con callbacks

**⏱️ Duración**: 90-120 minutos
