## Ejercicios con Mapas Interactivos y H3

```python
import plotly.express as px
import pandas as pd
import h3
import numpy as np

# Datos Mendoza
np.random.seed(42)
df = pd.DataFrame({
    'lat': np.random.normal(-32.889, 0.05, 500),
    'lon': np.random.normal(-68.846, 0.05, 500),
    'ventas': np.random.exponential(1000, 500)
})

# Ejercicio 1: Scatter Mapbox
fig = px.scatter_mapbox(df, lat='lat', lon='lon', size='ventas', 
                        mapbox_style='open-street-map', zoom=11)
fig.show()

# Ejercicio 2: Agregar H3
df['h3'] = df.apply(lambda r: h3.latlng_to_cell(r['lat'], r['lon'], 8), axis=1)
zonas = df.groupby('h3')['ventas'].sum().reset_index()
zonas['lat'] = zonas['h3'].apply(lambda x: h3.cell_to_latlng(x)[0])
zonas['lon'] = zonas['h3'].apply(lambda x: h3.cell_to_latlng(x)[1])

fig2 = px.scatter_mapbox(zonas, lat='lat', lon='lon', color='ventas',
                         mapbox_style='carto-positron', zoom=11)
fig2.show()

print("✅ Ejercicios completados")
```
