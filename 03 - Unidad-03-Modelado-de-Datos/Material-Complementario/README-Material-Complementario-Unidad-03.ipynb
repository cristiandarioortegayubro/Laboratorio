### 🌳 Módulo 01: Ensemble Methods Avanzados

**🎯 Objetivos:**
* Dominar métodos de ensemble (Bagging, Boosting, Stacking)
* Implementar Random Forest, XGBoost, LightGBM
* Optimizar hiperparámetros con Grid/Random Search
* Interpretar modelos complejos con SHAP

**📓 Notebooks:**

#### [01-Teoria-Ensemble-Methods-Avanzados](#notebook-1402451318137598)
* **Contenido**:
  * **Bagging (Bootstrap Aggregating)**:
    * Random Forest
    * Feature importance
    * Hiperparámetros clave
  * **Boosting**:
    * Gradient Boosting
    * XGBoost (Extreme Gradient Boosting)
    * LightGBM (más rápido que XGBoost)
    * Comparación de métodos
  * **Stacking**:
    * Meta-model con predicciones de nivel 0
    * StackingClassifier
  * **Optimización**:
    * Grid Search
    * Random Search (más eficiente)
  * **Interpretabilidad**:
    * SHAP (SHapley Additive exPlanations)
    * Summary plots, force plots

#### [01-Practica-Ensemble-Methods-Avanzados](#notebook-1402451318137599)
* **Ejercicios**:
  1. Random Forest para clasificación
  2. XGBoost con hiperparámetros optimizados
  3. LightGBM para grandes volúmenes
  4. Comparación de modelos (accuracy)
  5. Grid Search para tuning automático
* **Dataset**: Clasificación binaria (alta demanda)

**⏱️ Duración**: 90-120 minutos

#### [02-Teoria-Redes-Neuronales-Keras](#notebook-1402451318137600)

  * **Neurona Artificial**:

  * **Arquitecturas Avanzadas**:
    * CNN (Convolutional Neural Networks) para imágenes

  * Mejores prácticas

#### [02-Practica-Redes-Neuronales-Keras](#notebook-1402451318137601)

  1. Red neuronal simple (Dense + Dropout)

  4. Visualización de learning curves

### 🔍 Módulo 03: Detección de Anomalías

**🎯 Objetivos:**
* Identificar outliers y anomalías
* Aplicar métodos estadísticos (Z-score, IQR)
* Implementar algoritmos ML (Isolation Forest, One-Class SVM, LOF)
* Evaluar modelos no supervisados

**📓 Notebooks:**

#### [03-Teoria-Deteccion-Anomalias](#notebook-1402451318137602)
* **Contenido**:
  * **¿Qué es una Anomalía?**
    * Point anomaly, contextual, collective
    * Aplicaciones (fraude, seguridad, mantenimiento)
  * **Métodos Estadísticos**:
    * Z-Score (> 3 desviaciones estándar)
    * IQR (Interquartile Range)
  * **Métodos de Machine Learning**:
    * **Isolation Forest**: Anomalías fáciles de aislar
    * **One-Class SVM**: Frontera de normalidad
    * **LOF (Local Outlier Factor)**: Densidad local
  * **Evaluación**:
    * Con etiquetas (supervised)
    * Sin etiquetas (unsupervised)
    * Validación de negocio
  * Mejores prácticas

#### [03-Practica-Deteccion-Anomalias](#notebook-1402451318137603)
* **Ejercicios**:
  1. Isolation Forest para detección
  2. One-Class SVM
  3. Local Outlier Factor (LOF)
  4. Visualización comparativa (3 métodos)
* **Dataset**: Datos 2D con anomalías simuladas

**⏱️ Duración**: 75-90 minutos

---

### 📈 Módulo 04: Modelos de Series Temporales

**🎯 Objetivos:**
* Entender componentes de series temporales
* Aplicar modelos ARIMA, Prophet, LSTM
* Evaluar forecasts con métricas apropiadas
* Manejar estacionalidad y tendencias

**📓 Notebooks:**

#### [04-Teoria-Modelos-Series-Temporales](#notebook-1402451318137604)
* **Contenido**:
  * **Componentes de Series Temporales**:
    * Tendencia, estacionalidad, ruido
    * Descomposición aditiva/multiplicativa
  * **ARIMA (AutoRegressive Integrated Moving Average)**:
    * Componentes AR(p), I(d), MA(q)
    * Implementación con statsmodels
    * Auto ARIMA (selección automática)
  * **Prophet (Facebook)**:
    * Manejo de outliers y holidays
    * Múltiples estacionalidades
    * Robusto ante datos faltantes
  * **LSTM para Series Temporales**:
    * Ventana deslizante
    * Reshape para LSTM
    * Arquitectura secuencial
  * **Evaluación de Forecasts**:
    * MAE, RMSE, MAPE
    * Train/test split temporal (NO shuffle)
  * **Cuándo usar cada método**:
    * ARIMA: series simples, interpretable
    * Prophet: business forecasting
    * LSTM: patrones complejos
  * Mejores prácticas

#### [04-Practica-Modelos-Series-Temporales](#notebook-1402451318137605)
* **Ejercicios**:
  1. ARIMA para forecasting
  2. Prophet con estacionalidad anual
  3. Comparación visual de forecasts
  4. Métricas MAE y RMSE
* **Dataset**: Serie temporal sintética (3 años)

**⏱️ Duración**: 90-120 minutos
