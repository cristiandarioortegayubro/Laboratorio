# 💻 Módulo 03 - Práctica: A/B Testing

```python
import numpy as np
from scipy import stats
from statsmodels.stats.proportion import proportions_ztest
import pandas as pd

# Simular experimento A/B
np.random.seed(42)
n_users_A = 5000
n_users_B = 5000

# Grupo A (Control): 10% conversion
conversions_A = np.random.binomial(1, 0.10, n_users_A)

# Grupo B (Tratamiento): 11% conversion (1% lift)
conversions_B = np.random.binomial(1, 0.11, n_users_B)

print("✅ Experimento A/B simulado")
print(f"Grupo A: {conversions_A.sum()} conversiones de {n_users_A} usuarios ({conversions_A.mean():.2%})")
print(f"Grupo B: {conversions_B.sum()} conversiones de {n_users_B} usuarios ({conversions_B.mean():.2%})")

# Test de proporciones
counts = [conversions_B.sum(), conversions_A.sum()]
nobs = [n_users_B, n_users_A]

z_stat, p_value = proportions_ztest(counts, nobs)

print(f"\nZ-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("✅ B es significativamente mejor (rechazamos H0)")
    lift = (conversions_B.mean() - conversions_A.mean()) / conversions_A.mean()
    print(f"Lift: {lift:.2%}")
else:
    print("❌ No hay diferencia significativa")

print("\n✅ Práctica completada")
```
