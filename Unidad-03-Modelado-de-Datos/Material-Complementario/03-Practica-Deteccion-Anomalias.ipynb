## Ejercicios con Isolation Forest, One-Class SVM y LOF

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest
from sklearn.svm import OneClassSVM
from sklearn.neighbors import LocalOutlierFactor
import matplotlib.pyplot as plt

# Generar datos con anomalías
np.random.seed(42)
X_normal = np.random.randn(500, 2)
X_anomalies = np.random.uniform(low=-4, high=4, size=(50, 2))
X = np.vstack([X_normal, X_anomalies])
y_true = np.array([1]*500 + [-1]*50)  # 1=normal, -1=anomalía

# Ejercicio 1: Isolation Forest
iso_forest = IsolationForest(contamination=0.1, random_state=42)
y_pred_iso = iso_forest.fit_predict(X)

print("✅ Isolation Forest:")
print(f"Anomalías detectadas: {np.sum(y_pred_iso == -1)}")
print(f"Accuracy: {np.mean(y_pred_iso == y_true):.4f}")

# Ejercicio 2: One-Class SVM
svm = OneClassSVM(nu=0.1, gamma='auto')
y_pred_svm = svm.fit_predict(X)

print("\n✅ One-Class SVM:")
print(f"Anomalías detectadas: {np.sum(y_pred_svm == -1)}")
print(f"Accuracy: {np.mean(y_pred_svm == y_true):.4f}")

# Ejercicio 3: Local Outlier Factor
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.1)
y_pred_lof = lof.fit_predict(X)

print("\n✅ LOF:")
print(f"Anomalías detectadas: {np.sum(y_pred_lof == -1)}")
print(f"Accuracy: {np.mean(y_pred_lof == y_true):.4f}")

# Ejercicio 4: Visualización
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, y_pred, title in zip(axes, 
                              [y_pred_iso, y_pred_svm, y_pred_lof],
                              ['Isolation Forest', 'One-Class SVM', 'LOF']):
    ax.scatter(X[y_pred == 1, 0], X[y_pred == 1, 1], c='blue', s=20, label='Normal')
    ax.scatter(X[y_pred == -1, 0], X[y_pred == -1, 1], c='red', s=50, marker='x', label='Anomaly')
    ax.set_title(title)
    ax.legend()

plt.tight_layout()
display(plt.gcf())
plt.close()

print("\n✅ Ejercicios completados")
```
