# Generador de Datos - Panadería La Espiga Dorada

Este notebook genera datasets sintéticos para la **Panadería La Espiga Dorada**, un negocio ficticio de retail que opera en Mendoza, Argentina.

## Datasets generados:

1. **productos.csv**: Catálogo de productos (panes, facturas, tortas, bebidas)
2. **sucursales.csv**: Información de las 3 sucursales
3. **clientes.csv**: Base de datos de clientes frecuentes
4. **ventas.csv**: Transacciones de ventas (2 años de historia)

## Características del dataset:

* **Período**: 2 años (2024-2025)
* **Transacciones**: ~50,000 ventas
* **Productos**: 45 productos diferentes
* **Sucursales**: 3 locales en Mendoza con coordenadas GPS
* **Clientes**: 500 clientes frecuentes con direcciones georreferenciadas
* **Geolocalización**: Índices H3 para análisis espacial
* **Estacionalidad**: Mayor venta de facturas los fines de semana
* **Tendencias**: Crecimiento de productos saludables
* **Datos faltantes**: Algunos registros tienen valores nulos (realismo)

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random
import h3

# Configurar semilla para reproducibilidad
np.random.seed(42)
random.seed(42)

print("Librerías importadas correctamente")
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
print(f"H3 version: {h3.__version__}")
```

```python
# Dataset 1: PRODUCTOS
productos_data = {
    'producto_id': range(1, 46),
    'nombre': [
        # Panes (15)
        'Pan Francés', 'Pan Integral', 'Pan de Molde', 'Pan de Campo', 'Pan de Viena',
        'Pan Gallego', 'Baguette', 'Pan de Salvado', 'Pan de Centeno', 'Pan Ciabatta',
        'Pan de Ajo', 'Pan Pita', 'Pan Hamburguesa', 'Pan Pancho', 'Flauta',
        # Facturas (12)
        'Medialuna Manteca', 'Medialuna Grasa', 'Bola de Fraile', 'Canon', 'Vigilante',
        'Sacramento', 'Churros', 'Biscuit', 'Cremona', 'Palmerita', 'Fosforito', 'Facturas Surtidas',
        # Tortas (8)
        'Torta de Chocolate', 'Torta de Frutilla', 'Torta Rogel', 'Torta de Coco',
        'Lemon Pie', 'Torta Selva Negra', 'Cheesecake', 'Tiramisú',
        # Bebidas (5)
        'Café Espresso', 'Café con Leche', 'Té', 'Jugo Natural', 'Agua Mineral',
        # Otros (5)
        'Sandwich Miga', 'Empanadas', 'Pizza Porción', 'Alfajor', 'Galletitas'
    ],
    'categoria': (
        ['Pan'] * 15 + 
        ['Facturas'] * 12 + 
        ['Tortas'] * 8 + 
        ['Bebidas'] * 5 + 
        ['Otros'] * 5
    ),
    'precio_unitario': [
        # Panes
        800, 950, 1200, 1000, 850, 1100, 900, 1050, 1150, 950,
        750, 600, 500, 450, 800,
        # Facturas
        350, 280, 400, 350, 380, 380, 250, 320, 400, 330, 280, 3500,
        # Tortas
        8500, 9000, 9500, 8000, 7500, 10000, 9500, 11000,
        # Bebidas
        800, 950, 600, 1200, 500,
        # Otros
        2500, 800, 1500, 650, 900
    ],
    'costo_unitario': [
        # Panes (margen 50%)
        400, 475, 600, 500, 425, 550, 450, 525, 575, 475,
        375, 300, 250, 225, 400,
        # Facturas (margen 60%)
        140, 112, 160, 140, 152, 152, 100, 128, 160, 132, 112, 1400,
        # Tortas (margen 40%)
        5100, 5400, 5700, 4800, 4500, 6000, 5700, 6600,
        # Bebidas (margen 65%)
        280, 332, 210, 420, 175,
        # Otros (margen 55%)
        1125, 360, 675, 292, 405
    ],
    'stock_minimo': [
        # Panes
        50, 30, 20, 15, 40, 25, 35, 20, 15, 20,
        30, 25, 40, 40, 30,
        # Facturas
        100, 120, 50, 60, 40, 40, 80, 60, 30, 50, 60, 20,
        # Tortas
        5, 5, 4, 4, 3, 3, 4, 3,
        # Bebidas
        30, 40, 30, 20, 50,
        # Otros
        30, 40, 25, 50, 40
    ],
    'es_saludable': (
        [False, True, False, False, False, False, False, True, True, False,
         False, False, False, False, False] +  # Panes
        [False] * 12 +  # Facturas
        [False] * 8 +  # Tortas
        [False, False, True, True, True] +  # Bebidas
        [False] * 5  # Otros
    )
}

df_productos = pd.DataFrame(productos_data)
df_productos['margen_porcentaje'] = ((df_productos['precio_unitario'] - df_productos['costo_unitario']) / 
                                      df_productos['precio_unitario'] * 100).round(2)

print(f"\u2705 Productos generados: {len(df_productos)} productos")
df_productos.head(10)
```

```python
# Dataset 2: SUCURSALES
sucursales_data = {
    'sucursal_id': [1, 2, 3],
    'nombre': ['Espiga Dorada - Centro', 'Espiga Dorada - Guaymallén', 'Espiga Dorada - Godoy Cruz'],
    'direccion': ['Av. San Martín 1234, Mendoza', 'Ruta 40 Km 5, Guaymallén', 'Av. San Francisco de Asis 567, Godoy Cruz'],
    'zona': ['Centro', 'Este', 'Sur'],
    'latitud': [-32.8895, -32.9000, -32.9200],
    'longitud': [-68.8458, -68.8100, -68.8700],
    'fecha_apertura': ['2020-03-15', '2021-06-20', '2022-11-10'],
    'metros_cuadrados': [120, 150, 100],
    'empleados': [8, 10, 6],
    'tiene_cafeteria': [True, True, False]
}

df_sucursales = pd.DataFrame(sucursales_data)
df_sucursales['fecha_apertura'] = pd.to_datetime(df_sucursales['fecha_apertura'])

# Agregar índices H3 (resolución 9: ~0.1 km² por hexágono)
df_sucursales['h3_index'] = df_sucursales.apply(
    lambda row: h3.latlng_to_cell(row['latitud'], row['longitud'], 9), 
    axis=1
)

print(f"\u2705 Sucursales generadas: {len(df_sucursales)} sucursales")
print(f"   Con coordenadas GPS e índices H3")
df_sucursales
```

```python
# Dataset 3: CLIENTES
nombres = ['Juan', 'María', 'Carlos', 'Ana', 'Luis', 'Laura', 'Pedro', 'Sofía', 'Diego', 'Valentina',
           'Miguel', 'Florencia', 'Javier', 'Camila', 'Martín', 'Victoria', 'Alejandro', 'Martina', 'Gonzalo', 'Catalina']
apellidos = ['González', 'Rodríguez', 'Fernández', 'López', 'Martínez', 'Sánchez', 'Pérez', 'Gómez', 'Martín', 'Jiménez',
             'Ruiz', 'Díaz', 'Moreno', 'Álvarez', 'Romero', 'Alonso', 'Gutiérrez', 'Navarro', 'Torres', 'Domínguez']

# Generar coordenadas aleatorias en Mendoza (radio ~10km del centro)
centro_lat, centro_lon = -32.8895, -68.8458

clientes_data = []
for i in range(1, 501):
    nombre_completo = f"{random.choice(nombres)} {random.choice(apellidos)}"
    email = f"{nombre_completo.lower().replace(' ', '.')}@email.com" if random.random() > 0.1 else None
    telefono = f"261{random.randint(1000000, 9999999)}" if random.random() > 0.15 else None
    
    # Generar coordenadas aleatorias (distribución normal alrededor del centro)
    # Radio ~10km: 0.09 grados de latitud/longitud
    lat = centro_lat + np.random.normal(0, 0.03)
    lon = centro_lon + np.random.normal(0, 0.03)
    
    clientes_data.append({
        'cliente_id': i,
        'nombre': nombre_completo,
        'email': email,
        'telefono': telefono,
        'latitud': round(lat, 6),
        'longitud': round(lon, 6),
        'fecha_registro': (datetime(2020, 1, 1) + timedelta(days=random.randint(0, 1825))).strftime('%Y-%m-%d'),
        'preferencia_categoria': random.choice(['Pan', 'Facturas', 'Tortas', 'Bebidas', 'Otros', None]),
        'es_vip': random.random() > 0.85
    })

df_clientes = pd.DataFrame(clientes_data)
df_clientes['fecha_registro'] = pd.to_datetime(df_clientes['fecha_registro'])

# Agregar índices H3 para clientes (resolución 9)
df_clientes['h3_index'] = df_clientes.apply(
    lambda row: h3.latlng_to_cell(row['latitud'], row['longitud'], 9), 
    axis=1
)

print(f"\u2705 Clientes generados: {len(df_clientes)} clientes")
print(f"   Clientes VIP: {df_clientes['es_vip'].sum()}")
print(f"   Clientes con email: {df_clientes['email'].notna().sum()}")
print(f"   Clientes con teléfono: {df_clientes['telefono'].notna().sum()}")
print(f"   Con coordenadas GPS e índices H3 para análisis geoespacial")
df_clientes.head(10)
```

```python
# Dataset 4: VENTAS (2 años de historia)
print("Generando dataset de ventas... esto puede tomar unos segundos")

ventas_data = []
venta_id = 1
fecha_inicio = datetime(2024, 1, 1)
fecha_fin = datetime(2025, 12, 31)

# Generar ventas para cada día
fecha_actual = fecha_inicio
while fecha_actual <= fecha_fin:
    # Determinar número de ventas según día de semana
    es_fin_de_semana = fecha_actual.weekday() >= 5
    es_lunes = fecha_actual.weekday() == 0
    
    # Más ventas los fines de semana, menos los lunes
    if es_fin_de_semana:
        num_ventas = random.randint(80, 120)
    elif es_lunes:
        num_ventas = random.randint(40, 60)
    else:
        num_ventas = random.randint(60, 85)
    
    # Generar ventas del día
    for _ in range(num_ventas):
        sucursal_id = random.choice([1, 2, 3])
        
        # 70% de las ventas tienen cliente identificado
        cliente_id = random.choice(df_clientes['cliente_id'].tolist()) if random.random() > 0.3 else None
        
        # Hora de venta (picos en desayuno y merienda)
        hora_pico = random.choice(['desayuno', 'almuerzo', 'merienda'])
        if hora_pico == 'desayuno':
            hora = random.randint(7, 10)
        elif hora_pico == 'almuerzo':
            hora = random.randint(12, 14)
        else:  # merienda
            hora = random.randint(17, 20)
        
        minuto = random.randint(0, 59)
        segundo = random.randint(0, 59)
        fecha_hora = fecha_actual.replace(hour=hora, minute=minuto, second=segundo)
        
        ventas_data.append({
            'venta_id': venta_id,
            'fecha': fecha_actual.strftime('%Y-%m-%d'),
            'hora': fecha_hora.strftime('%H:%M:%S'),
            'sucursal_id': sucursal_id,
            'cliente_id': cliente_id
        })
        venta_id += 1
    
    fecha_actual += timedelta(days=1)
    
    # Mostrar progreso cada 100 días
    if (fecha_actual - fecha_inicio).days % 100 == 0:
        print(f"  Progreso: {(fecha_actual - fecha_inicio).days} días procesados...")

print(f"\n\u2705 Ventas generadas: {len(ventas_data)} transacciones")
```

```python
# Generar detalles de cada venta (productos comprados)
print("Generando detalles de ventas...")
detalles_ventas = []

for venta in ventas_data:
    # Cada venta tiene entre 1 y 5 productos diferentes
    num_productos = random.choices([1, 2, 3, 4, 5], weights=[40, 30, 20, 7, 3])[0]
    
    # Seleccionar productos aleatorios
    productos_venta = random.sample(df_productos['producto_id'].tolist(), num_productos)
    
    for producto_id in productos_venta:
        cantidad = random.choices([1, 2, 3, 4, 6, 12], weights=[50, 25, 15, 5, 3, 2])[0]
        
        # Buscar precio del producto
        precio = df_productos[df_productos['producto_id'] == producto_id]['precio_unitario'].values[0]
        
        # 5% de descuento aleatorio en algunos productos
        descuento = random.choice([0, 0, 0, 0, 0, 0, 0, 0, 0, 5, 10, 15])
        
        detalles_ventas.append({
            'venta_id': venta['venta_id'],
            'producto_id': producto_id,
            'cantidad': cantidad,
            'precio_unitario': precio,
            'descuento_porcentaje': descuento,
            'subtotal': precio * cantidad * (1 - descuento/100)
        })

print(f"\u2705 Detalles de ventas generados: {len(detalles_ventas)} líneas de venta")
print(f"   Promedio de productos por venta: {len(detalles_ventas) / len(ventas_data):.2f}")
```

```python
# Crear DataFrames finales
df_ventas = pd.DataFrame(ventas_data)
df_detalles_ventas = pd.DataFrame(detalles_ventas)

# Agregar total a cada venta
df_totales = df_detalles_ventas.groupby('venta_id')['subtotal'].sum().reset_index()
df_totales.columns = ['venta_id', 'total']
df_ventas = df_ventas.merge(df_totales, on='venta_id')

# Convertir fechas
df_ventas['fecha'] = pd.to_datetime(df_ventas['fecha'])
df_ventas['fecha_hora'] = pd.to_datetime(df_ventas['fecha'].astype(str) + ' ' + df_ventas['hora'])

# Agregar columnas adicionales
df_ventas['dia_semana'] = df_ventas['fecha'].dt.day_name()
df_ventas['mes'] = df_ventas['fecha'].dt.month
df_ventas['anio'] = df_ventas['fecha'].dt.year
df_ventas['es_fin_de_semana'] = df_ventas['fecha'].dt.weekday >= 5

print("\n\u2705 DataFrames finales creados")
print(f"\nEstadísticas de ventas:")
print(f"  Total ventas: {len(df_ventas):,}")
print(f"  Total líneas de venta: {len(df_detalles_ventas):,}")
print(f"  Facturación total: ${df_ventas['total'].sum():,.2f}")
print(f"  Ticket promedio: ${df_ventas['total'].mean():,.2f}")
print(f"  Venta mínima: ${df_ventas['total'].min():,.2f}")
print(f"  Venta máxima: ${df_ventas['total'].max():,.2f}")
```

## 🗺️ Capacidades Geoespaciales con H3

Este dataset ahora incluye **análisis geoespacial** utilizando el sistema de indexación **H3** de Uber:

### ¿Qué es H3?
H3 es un sistema de cuadrícula geoespacial jerárquico que divide la Tierra en hexágonos. Cada ubicación GPS se convierte en un índice H3 único, facilitando:
- **Agregaciones espaciales** (agrupar datos por zonas)
- **Cálculo de distancias** entre ubicaciones
- **Heatmaps** de densidad
- **Optimización de rutas** y áreas de cobertura

### Datos Georreferenciados Generados

1. **Sucursales** (`sucursales.csv`)
   - Coordenadas GPS reales de Mendoza, Argentina
   - Índices H3 (resolución 9: ~0.1 km²)
   - 3 sucursales con ubicaciones estratégicas

2. **Clientes** (`clientes.csv`)
   - 500 clientes con direcciones georreferenciadas
   - Distribución normal alrededor del centro de Mendoza (~10km radio)
   - Índices H3 para análisis de densidad

3. **Análisis Disponibles**
   - Distancias entre sucursales (en hexágonos)
   - Clientes por zona (hexágonos H3)
   - Área de cobertura de cada sucursal
   - Facturación por zona geográfica
   - Top hexágonos con mayor demanda

### Casos de Uso para el Laboratorio

✅ **Unidad 1 - Análisis de Datos**: Explorar patrones geográficos en ventas  
✅ **Unidad 2 - Visualización**: Crear heatmaps y mapas de densidad  
✅ **Unidad 3 - Modelado**: Feature engineering espacial (distancia a sucursal más cercana)  
✅ **Unidad 4 - Proyectos**: Optimización de rutas de delivery, selección de nuevas ubicaciones

```python
print("=" * 80)
print("ANÁLISIS GEOESPACIAL CON H3")
print("=" * 80)

# 1. Información de las sucursales
print("\n1. Coordenadas de Sucursales:")
print(df_sucursales[['nombre', 'zona', 'latitud', 'longitud', 'h3_index']])

# 2. Distancias entre sucursales usando H3
print("\n2. Distancias entre Sucursales (en número de hexágonos H3):")
sucursales_list = df_sucursales[['sucursal_id', 'nombre', 'h3_index']].to_dict('records')

for i, suc1 in enumerate(sucursales_list):
    for suc2 in sucursales_list[i+1:]:
        distancia = h3.grid_distance(suc1['h3_index'], suc2['h3_index'])
        print(f"   {suc1['nombre']} <-> {suc2['nombre']}: {distancia} hexágonos")

# 3. Clientes por hexágono H3
print("\n3. Top 10 Hexágonos H3 con Más Clientes:")
clientes_por_h3 = df_clientes['h3_index'].value_counts().head(10)
print(clientes_por_h3)

# 4. Número único de hexágonos ocupados
print(f"\n4. Cobertura Geográfica:")
print(f"   Total de hexágonos únicos con clientes: {df_clientes['h3_index'].nunique()}")
print(f"   Total de clientes: {len(df_clientes)}")
print(f"   Promedio de clientes por hexágono: {len(df_clientes) / df_clientes['h3_index'].nunique():.2f}")
```

```python
print("=" * 80)
print("ÁREA DE COBERTURA DE SUCURSALES (HEXÁGONOS VECINOS)")
print("=" * 80)

# Para cada sucursal, obtener hexágonos vecinos (radio 2)
for idx, row in df_sucursales.iterrows():
    h3_index = row['h3_index']
    nombre = row['nombre']
    
    # Obtener hexágonos vecinos en un radio de 2
    vecinos = h3.grid_disk(h3_index, 2)
    
    # Contar cuántos clientes hay en el área de cobertura
    clientes_cercanos = df_clientes[df_clientes['h3_index'].isin(vecinos)]
    
    print(f"\n{nombre}:")
    print(f"   Índice H3: {h3_index}")
    print(f"   Hexágonos en área de cobertura (radio 2): {len(vecinos)}")
    print(f"   Clientes en área de cobertura: {len(clientes_cercanos)} ({len(clientes_cercanos)/len(df_clientes)*100:.1f}%)")
```

```python
print("=" * 80)
print("EJEMPLO: ANÁLISIS DE VENTAS POR ZONA GEOGRÁFICA")
print("=" * 80)

# Unir ventas con clientes para obtener h3_index
ventas_con_geo = df_ventas[df_ventas['cliente_id'].notna()].merge(
    df_clientes[['cliente_id', 'h3_index']], 
    on='cliente_id'
)

print(f"\nVentas con información geográfica: {len(ventas_con_geo):,} ({len(ventas_con_geo)/len(df_ventas)*100:.1f}%)")

# Top 10 hexágonos por facturación
print("\nTop 10 Hexágonos por Facturación:")
ventas_por_h3 = ventas_con_geo.groupby('h3_index').agg({
    'total': ['sum', 'count', 'mean']
}).round(2)
ventas_por_h3.columns = ['Facturación Total', 'Nº Ventas', 'Ticket Promedio']
ventas_por_h3 = ventas_por_h3.sort_values('Facturación Total', ascending=False).head(10)
print(ventas_por_h3)

print("\nℹ️ Casos de uso con H3:")
print("   - Identificar zonas de alta demanda para abrir nuevas sucursales")
print("   - Optimizar rutas de delivery por proximidad geográfica")
print("   - Crear heatmaps de facturación por zona")
print("   - Segmentar clientes por ubicación geográfica")
print("   - Analizar alcance de campañas de marketing por zona")
```
