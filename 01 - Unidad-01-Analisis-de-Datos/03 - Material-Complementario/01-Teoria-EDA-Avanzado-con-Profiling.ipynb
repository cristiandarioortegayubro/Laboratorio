## 1️⃣ Introducción al EDA Avanzado

### ¿Qué es EDA (Exploratory Data Analysis)?

El **Análisis Exploratorio de Datos** es el proceso de:
* 🔍 **Examinar** la estructura y contenido de los datos
* 📊 **Visualizar** distribuciones y relaciones
* ⚠️ **Detectar** anomalías, outliers y problemas de calidad
* 💡 **Generar hipótesis** para análisis posterior
* 🎯 **Informar** decisiones de preprocesamiento y modelado

### EDA Básico vs EDA Avanzado

| Aspecto | EDA Básico | EDA Avanzado |
|---------|------------|---------------|
| **Herramientas** | `.info()`, `.describe()`, `.value_counts()` | Profiling automatizado, reportes HTML |
| **Tiempo** | Manual, 30-60 min | Automatizado, 2-5 min |
| **Profundidad** | Estadísticas básicas | Correlaciones, interacciones, distribuciones detalladas |
| **Visualización** | Gráficos básicos | Dashboards interactivos |
| **Detección** | Manual | Automática (missing, duplicados, outliers, alertas) |
| **Comparación** | Difícil | Comparación side-by-side de datasets |

### ¿Por qué EDA Avanzado?

✅ **Ahorro de tiempo**: 30-60 minutos → 2-5 minutos  
✅ **Completitud**: Nunca te pierdes un análisis importante  
✅ **Reproducibilidad**: Genera el mismo reporte siempre  
✅ **Comunicación**: Reportes HTML compartibles con stakeholders  
✅ **Escalabilidad**: Funciona igual para 10 o 10,000 columnas  
✅ **Automatización**: Integrable en pipelines de CI/CD

## 2️⃣ Herramientas de Profiling Automatizado

### Principales Librerías

Existen varias librerías Python para EDA automatizado:

#### 🐍 **ydata-profiling** (anteriormente pandas-profiling)
* 🎯 **Uso**: EDA completo de un solo dataset
* 📄 **Output**: Reporte HTML interactivo
* 📊 **Features**: Distribuciones, correlaciones, missing values, duplicados, warnings
* 📖 **Docs**: [ydata-profiling.ydata.ai](https://docs.profiling.ydata.ai/)

#### 🍬 **Sweetviz**
* 🎯 **Uso**: EDA + comparación de datasets (train vs test)
* 📄 **Output**: Reporte HTML con visualizaciones comparativas
* 📊 **Features**: Análisis univariado, bivariado, target analysis
* 📖 **Docs**: [github.com/fbdesignpro/sweetviz](https://github.com/fbdesignpro/sweetviz)

#### Otras Herramientas

**D-Tale**:
* Dashboard interactivo estilo Excel
* Edición de datos en vivo
* Ideal para exploración interactiva

**AutoViz**:
* Visualizaciones automáticas
* Enfocado en gráficos estáticos
* Rápido para datasets grandes

**DataPrep**:
* Pipeline completo (EDA + limpieza)
* Integración con ML
* Soporte para Big Data

### Uso Básico

```python
import pandas as pd
from ydata_profiling import ProfileReport

# Cargar datos
df = pd.read_csv("/dbfs/FileStore/ventas.csv")

# Generar reporte
profile = ProfileReport(df, title="Análisis de Ventas")
profile.to_file("reporte_ventas.html")
```

### Sección del Reporte

El reporte incluye:

#### **1. Overview**
* Número de variables, observaciones
* Tipos de datos
* Missing values (%)
* Duplicados
* Memoria usada
* Warnings automáticos

#### **2. Variables (por columna)**
* **Numéricas**: Distribuciones, histogramas, estadísticas (mean, std, min, max, quantiles)
* **Categóricas**: Frecuencias, gráficos de barras, categorías únicas
* **Fechas**: Rango, frecuencia temporal
* **Texto**: Longitud, patrones

#### **3. Interacciones**
* **Correlaciones**: Pearson, Spearman, Kendall
* **Matriz de correlación**: Heatmap interactivo
* **Missing values**: Patrones de valores faltantes

#### **4. Alerts**
* Alta cardinalidad
* Columnas constantes
* Alta correlación (posible multicolinealidad)
* Muchos ceros
* Skewness extremo

### Configuración Avanzada

```python
profile = ProfileReport(
    df,
    title="Ventas Panadería",
    minimal=False,  # Reporte completo
    explorative=True,  # Más visualizaciones
    dark_mode=False,
    correlations={
        "pearson": {"calculate": True},
        "spearman": {"calculate": True},
        "kendall": {"calculate": False},  # Lento
        "phi_k": {"calculate": False}
    }
)
```

### Integración con Notebooks

```python
# Ver reporte directamente en el notebook (no HTML)
profile.to_notebook_iframe()

# O como widget interactivo
profile.to_widgets()
```

## 5️⃣ Comparación de Herramientas

### ¿Cuándo usar cada una?

#### 🔵 **ydata-profiling**

**Usar cuando:**
* 👉 Necesitas un EDA **completo y exhaustivo**
* 👉 Dataset nuevo que no conoces
* 👉 Quieres **detectar todos los problemas** posibles
* 👉 Necesitas **documentación detallada** del dataset
* 👉 Dataset no es demasiado grande (<100K filas ideal)

**Ejemplo:**
* Primer análisis de datos de cliente nuevo
* Auditoría de calidad de datos
* Documentación de datasets para equipo

---

#### 🟢 **Sweetviz**

**Usar cuando:**
* 👉 Necesitas **comparar dos datasets** (train/test, before/after)
* 👉 Tienes una **variable objetivo** (clasificación, regresión)
* 👉 Quieres análisis **rápido** en datasets grandes
* 👉 Necesitas ver **asociaciones con target**

**Ejemplo:**
* Validar train/test split
* Analizar features vs target antes de modelar
* Comparar datos pre y post limpieza

---

### 💡 Estrategia Recomendada

**Workflow típico:**

1. **Paso 1**: `ydata-profiling` en dataset crudo → Entender todo
2. **Paso 2**: Limpieza y feature engineering basado en insights
3. **Paso 3**: `sweetviz` para comparar train/test y analizar target
4. **Paso 4**: Modelado

**Tiempo total**: 10-15 minutos de EDA automático vs 2-3 horas manual

## 6️⃣ Mejores Prácticas

### ✅ **Qué Hacer**

1. **Ejecutar profiling ANTES de cualquier transformación**
   * Entérate de problemas en datos crudos
   * Documenta estado original

2. **Configurar profiling según tamaño del dataset**
   * Datasets grandes (>100K filas): `minimal=True` o sweetviz
   * Datasets pequeños: `explorative=True` en ydata-profiling

3. **Guardar reportes HTML con timestamp**
   ```python
   from datetime import datetime
   timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
   profile.to_file(f"eda_ventas_{timestamp}.html")
   ```

4. **Revisar warnings automáticamente**
   * Alta correlación → Multicolinealidad
   * Alta cardinalidad → Posible ID o dato único
   * Muchos zeros → Sparse data

5. **Comparar datasets con Sweetviz**
   * Train vs Test → Detectar data leakage o distribuciones diferentes
   * Before vs After limpieza → Validar transformaciones

6. **Integrar en pipelines CI/CD**
   ```python
   # En pipeline de datos
   profile = ProfileReport(df, title="Data Quality Check")
   
   # Extraer métricas
   if profile.get_description()["table"]["n_duplicates"] > 100:
       raise ValueError("Demasiados duplicados!")
   ```

---

### ❌ **Qué NO Hacer**

1. **NO ejecutar profiling en datos ya transformados sin guardar original**
   * Pierdes visibilidad de problemas originales

2. **NO ignorar warnings**
   * Son señales de posibles problemas

3. **NO usar profiling como único EDA**
   * Complementa con visualizaciones específicas

4. **NO profiling en producción**
   * Solo en desarrollo/análisis
   * Consume memoria y tiempo

5. **NO compartir reportes con datos sensibles**
   * Revisar antes de compartir externamente
   * Considerar anonimización
