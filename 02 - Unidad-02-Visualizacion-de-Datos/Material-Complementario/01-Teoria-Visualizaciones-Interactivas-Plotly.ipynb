## 1️⃣ Introducción a Plotly

### ¿Qué es Plotly?

**Plotly** es una librería de visualización interactiva basada en JavaScript (Plotly.js) con API en Python.

**Ventajas vs Matplotlib/Seaborn:**
* ✅ **Interactividad nativa**: Zoom, pan, hover, selección
* ✅ **Calidad profesional**: Gráficos listos para publicación
* ✅ **Dashboards fáciles**: Dash framework integrado
* ✅ **Export web**: HTML standalone
* ✅ **Responsive**: Se adapta a pantallas

### Arquitectura de Plotly

```
Plotly Ecosystem
├── Plotly Express (px)    ← API de alto nivel (recomendado)
├── Graph Objects (go)      ← API de bajo nivel (control total)
├── Dash                    ← Framework de dashboards
└── Plotly.js              ← Motor JavaScript (usado internamente)
```

### Tipos de Gráficos

#### **1. Scatter Plot (Dispersión)**

```python
fig = px.scatter(
    df, 
    x='precio', 
    y='ventas',
    color='categoria',          # Color por grupo
    size='cantidad',            # Tamaño por variable
    hover_data=['producto'],    # Info en tooltip
    title='Ventas vs Precio por Categoría'
)
fig.show()
```

**💡 Interactividad:**
* Hover: Ver datos del punto
* Zoom: Box select o scroll
* Pan: Arrastrar gráfico
* Doble click: Reset zoom

#### **2. Line Chart (Líneas)**

```python
fig = px.line(
    df_ts, 
    x='fecha', 
    y='ingresos',
    color='sucursal',
    title='Evolución de Ingresos'
)
fig.update_traces(mode='lines+markers')  # Línea + puntos
fig.show()
```

#### **3. Bar Chart (Barras)**

```python
fig = px.bar(
    df_agg,
    x='mes',
    y='ventas',
    color='categoria',
    barmode='group',  # 'stack' para apilado
    title='Ventas Mensuales por Categoría'
)
fig.show()
```

#### **4. Histograma**

```python
fig = px.histogram(
    df,
    x='monto',
    nbins=50,
    color='segmento',
    marginal='box',  # Boxplot en margen
    title='Distribución de Montos'
)
fig.show()
```

#### **5. Box Plot**

```python
fig = px.box(
    df,
    x='categoria',
    y='precio',
    color='categoria',
    points='all',  # Mostrar todos los puntos
    title='Distribución de Precios'
)
fig.show()
```

#### **6. Heatmap (Mapa de Calor)**

```python
corr_matrix = df.corr()

fig = px.imshow(
    corr_matrix,
    text_auto=True,  # Valores en celdas
    aspect='auto',
    color_continuous_scale='RdBu_r',  # Paleta
    title='Matriz de Correlación'
)
fig.show()
```

#### **7. Sunburst (Jerárquico)**

```python
fig = px.sunburst(
    df,
    path=['region', 'ciudad', 'sucursal'],  # Jerarquía
    values='ventas',
    color='margen_pct',
    title='Ventas por Ubicación'
)
fig.show()
```

## 3️⃣ Personalización Avanzada

### Update Layout (Diseño)

```python
fig.update_layout(
    title={
        'text': 'Título Centrado',
        'x': 0.5,
        'xanchor': 'center'
    },
    xaxis_title='Eje X (£)',
    yaxis_title='Eje Y (unidades)',
    font=dict(size=14),
    template='plotly_white',  # Tema
    hovermode='x unified',     # Hover alineado
    showlegend=True,
    legend=dict(
        orientation='h',
        yanchor='bottom',
        y=1.02
    )
)
```

### Temas Disponibles

```python
import plotly.io as pio
print(pio.templates)

# Temas populares:
# - plotly (default)
# - plotly_white
# - plotly_dark
# - seaborn
# - ggplot2
# - simple_white
```

### Anotaciones

```python
fig.add_annotation(
    x=10, y=20,
    text='Outlier importante',
    showarrow=True,
    arrowhead=2,
    ax=-40, ay=-40
)
```

### Líneas de Referencia

```python
# Línea horizontal (umbral)
fig.add_hline(
    y=100,
    line_dash='dash',
    line_color='red',
    annotation_text='Objetivo'
)

# Línea vertical (evento)
fig.add_vline(
    x='2024-03-01',
    line_dash='dot',
    line_color='green',
    annotation_text='Lanzamiento'
)
```

## 4️⃣ Subplots y Layouts Complejos

### Subplots (Grid)

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Ventas', 'Clientes', 'Margen', 'Inventario'),
    specs=[[{'type': 'scatter'}, {'type': 'bar'}],
           [{'type': 'pie'}, {'type': 'indicator'}]]
)

# Añadir trazas
fig.add_trace(go.Scatter(x=[1,2,3], y=[4,5,6]), row=1, col=1)
fig.add_trace(go.Bar(x=['A','B'], y=[10,20]), row=1, col=2)
fig.add_trace(go.Pie(labels=['X','Y'], values=[30,70]), row=2, col=1)
fig.add_trace(go.Indicator(
    mode='number+delta',
    value=150,
    delta={'reference': 100}
), row=2, col=2)

fig.update_layout(height=600, showlegend=False)
fig.show()
```

## 5️⃣ Animaciones

### Animación Temporal

```python
fig = px.scatter(
    df,
    x='x',
    y='y',
    animation_frame='anio',  # Dimensión temporal
    animation_group='pais',  # Agrupar animación
    size='poblacion',
    color='continente',
    hover_name='pais',
    range_x=[0, 100],
    range_y=[0, 100],
    title='Evolución Temporal'
)

fig.layout.updatemenus[0].buttons[0].args[1]['frame']['duration'] = 1000  # ms
fig.show()
```

## ✅ Resumen

### Cuándo Usar Plotly

✅ **Usar Plotly cuando:**
* Visualizaciones interactivas para web
* Dashboards dinámicos
* Presentaciones ejecutivas
* Compartir gráficos standalone

❌ **No usar Plotly cuando:**
* Gráficos rápidos exploratorios (usar matplotlib)
* PDFs estáticos sin interactividad
* Publicaciones académicas (preferir matplotlib)

### Tipos de Gráficos por Caso de Uso

| Caso de Uso | Tipo de Gráfico |
|-------------|------------------|
| Tendencias temporales | Line, Area |
| Comparaciones | Bar, Column |
| Distribuciones | Histogram, Box, Violin |
| Correlaciones | Scatter, Heatmap |
| Proporciones | Pie, Sunburst, Treemap |
| Geoespacial | Scatter_mapbox, Choropleth |
| KPIs | Indicator, Gauge |
