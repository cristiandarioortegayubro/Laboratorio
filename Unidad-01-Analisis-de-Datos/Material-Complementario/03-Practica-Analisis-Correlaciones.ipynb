## 🛠️ Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import spearmanr, kendalltau
import warnings
warnings.filterwarnings('ignore')

# Cargar datos
ventas = pd.read_csv("/dbfs/FileStore/Laboratorio/Datasets/ventas.csv").head(1000)
print("✅ Datos cargados")
```

---

## 📊 Ejercicio 1: Matriz de Correlación

```python
# Calcular matriz de correlación (Pearson)
corr_matrix = ventas.select_dtypes(include=[np.number]).corr()

# Visualizar con heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title("Matriz de Correlación (Pearson)")
plt.tight_layout()
plt.show()

print("\n📊 Correlaciones más fuertes con 'monto':")
print(corr_matrix['monto'].abs().sort_values(ascending=False).head(6))
```

---

## 🔄 Ejercicio 2: Pearson vs Spearman

```python
# Comparar métodos
variables = ['monto', 'cantidad', 'precio']

print("📊 Comparación Pearson vs Spearman:\n")
for var1, var2 in [('monto', 'cantidad'), ('monto', 'precio')]:
    pearson = ventas[var1].corr(ventas[var2], method='pearson')
    spearman = ventas[var1].corr(ventas[var2], method='spearman')
    
    print(f"{var1} vs {var2}:")
    print(f"  Pearson:  {pearson:.3f}")
    print(f"  Spearman: {spearman:.3f}")
    print(f"  Diferencia: {abs(pearson - spearman):.3f}\n")

print("💡 Si Pearson y Spearman difieren mucho → Relación no lineal u outliers")
```

---

## ⚠️ Ejercicio 3: Detección de Multicolinealidad

```python
# Detectar pares altamente correlacionados
threshold = 0.8
corr_matrix_abs = corr_matrix.abs()

# Obtener pares (triangulo superior)
upper_triangle = corr_matrix_abs.where(np.triu(np.ones(corr_matrix_abs.shape), k=1).astype(bool))

high_corr_pairs = [(col, row) for col in upper_triangle.columns 
                   for row in upper_triangle.index 
                   if upper_triangle.loc[row, col] > threshold]

if high_corr_pairs:
    print(f"⚠️ Pares con correlación > {threshold}:")
    for var1, var2 in high_corr_pairs:
        corr_val = corr_matrix.loc[var2, var1]
        print(f"  {var1} ↔ {var2}: {corr_val:.3f}")
    print("\n💡 Acción: Eliminar una variable de cada par")
else:
    print("✅ Sin multicolinealidad fuerte detectada")
```

---

## 🎯 Ejercicio 4: Selección de Features

```python
# Crear target dummy
ventas['target'] = (ventas['monto'] > ventas['monto'].median()).astype(int)

# Paso 1: Correlación con target
features = ventas.select_dtypes(include=[np.number]).columns.drop('target')
target_corr = ventas[features].corrwith(ventas['target']).abs().sort_values(ascending=False)

print("🎯 Top 5 Features más correlacionados con target:\n")
for feature, corr in target_corr.head(5).items():
    icon = "🟢" if corr > 0.3 else "🟡" if corr > 0.1 else "⚪"
    print(f"{icon} {feature:20s} = {corr:.3f}")

# Paso 2: Seleccionar features
threshold_target = 0.1
selected_features = target_corr[target_corr > threshold_target].index.tolist()

print(f"\n✅ Features seleccionados ({len(selected_features)}): {selected_features}")
print(f"\n📊 Reducción: {len(features)} → {len(selected_features)} features")
```
