# 📚 Módulo 03 - Storytelling con Datos

## 🎯 Objetivos

1. ✅ Comunicar insights efectivamente
2. ✅ Estructurar narrativas con datos
3. ✅ Elegir visualizaciones apropiadas
4. ✅ Diseñar presentaciones ejecutivas

---

## 1️⃣ Principios de Storytelling

### Estructura de una Historia de Datos

**1. Contexto** (¿Por qué es importante?)
* Situación actual
* Problema u oportunidad
* Stakeholders afectados

**2. Complicación** (¿Qué está pasando?)
* Datos que evidencian el problema
* Tendencias preocupantes
* Comparaciones relevantes

**3. Resolución** (¿Qué hacer?)
* Insights accionables
* Recomendaciones específicas
* Impacto esperado

---

## 2️⃣ Elegir la Visualización Correcta

| Objetivo | Visualización |
|----------|----------------|
| Tendencia temporal | Line chart |
| Comparación | Bar chart |
| Distribución | Histogram, Box plot |
| Correlación | Scatter plot |
| Proporción | Pie, Sunburst |
| Geoespacial | Mapa, Choropleth |
| KPI único | Indicator, Gauge |

## 3️⃣ Principios de Diseño

### Ley de Gestalt

**Proximidad**: Elementos cercanos se perciben juntos  
**Similitud**: Elementos similares se agrupan mentalmente  
**Continuidad**: El ojo sigue líneas continuas  
**Cierre**: Completamos patrones incompletos

### Jerarquía Visual

1. **Primario**: Título, insight principal
2. **Secundario**: Subtextos, ejes
3. **Terciario**: Leyendas, anotaciones

### Color

✅ **Usar color con propósito**:
* Categorizar (cualitativo)
* Cuantificar (secuencial)
* Enfatizar (destacar)

❌ **Evitar**:
* > 5 colores diferentes
* Solo rojo-verde (daltonismo)
* Arcoiris (dificulta comparación)

---

## 4️⃣ Framework: Presentación Ejecutiva

### Slide 1: Resumen Ejecutivo
* 1 número grande (KPI principal)
* Δ% vs periodo anterior
* 1 frase de contexto

### Slide 2: Tendencias
* Line chart principal
* Anotaciones en eventos clave
* Pronóstico (si aplica)

### Slide 3: Segmentación
* Comparación por segmentos
* Top/Bottom performers
* Oportunidades identificadas

### Slide 4: Recomendaciones
* 3 acciones priorizadas
* Impacto estimado
* Next steps

---

## ✅ Checklist de Calidad

- [ ] Título descriptivo (no genérico)
- [ ] Ejes con unidades claras
- [ ] Colores con propósito
- [ ] Fuente legible (min 12pt)
- [ ] Contexto suficiente
- [ ] Insight evidente (no oculto)
- [ ] Recomendación accionable
