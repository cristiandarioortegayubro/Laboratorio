## 1️⃣ ¿Qué es una Anomalía?

**Anomalía** = Observación que se desvía significativamente del patrón normal

### Tipos

* **Point anomaly**: Un punto individual es anómalo
* **Contextual anomaly**: Normal en un contexto, anómalo en otro
* **Collective anomaly**: Secuencia de puntos juntos forman anomalía

### Aplicaciones

* 🛡️ Detección de fraude
* 📡 Seguridad de redes
* 🏭 Mantenimiento predictivo
* 📊 Calidad de datos

## 3️⃣ Métodos de Machine Learning

### Isolation Forest

**Idea**: Anomalías son más fáciles de aislar (menos particiones)

```python
from sklearn.ensemble import IsolationForest

iso_forest = IsolationForest(
    contamination=0.1,  # % esperado de anomalías
    random_state=42
)

y_pred = iso_forest.fit_predict(X)
anomalies = y_pred == -1  # -1 = anomalía, 1 = normal
```

**Ventajas:**
* Rápido
* Maneja datos de alta dimensión
* No requiere etiquetas

### One-Class SVM

```python
from sklearn.svm import OneClassSVM

svm = OneClassSVM(nu=0.1, gamma='auto')
y_pred = svm.fit_predict(X)
anomalies = y_pred == -1
```

**Parámetro `nu`**: % esperado de anomalías

### Local Outlier Factor (LOF)

```python
from sklearn.neighbors import LocalOutlierFactor

lof = LocalOutlierFactor(n_neighbors=20, contamination=0.1)
y_pred = lof.fit_predict(X)
anomalies = y_pred == -1
```

**Idea**: Anomalía tiene densidad local significativamente menor
