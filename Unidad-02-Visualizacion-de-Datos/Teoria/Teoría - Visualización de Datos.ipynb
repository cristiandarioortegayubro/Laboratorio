## 1️⃣ Principios de Visualización de Datos

### ¿Por qué visualizar datos?

**Una imagen vale más que mil números:**

1. **🧠 Comprensión rápida**: El cerebro procesa imágenes 60,000× más rápido que texto
2. **🔍 Detección de patrones**: Identifica tendencias, outliers, correlaciones
3. **💬 Comunicación efectiva**: Transmite insights a audiencias técnicas y no técnicas
4. **🎯 Toma de decisiones**: Apoya decisiones basadas en evidencia visual

### Principios Fundamentales

#### 1. **Claridad sobre Complejidad**
* Visualización simple > Visualización sofisticada confusa
* Elimina elementos innecesarios (chart junk)
* Enfócate en el mensaje principal

#### 2. **Precisión**
* No distorsiones visuales engañosas
* Escalas apropiadas (evita ejes truncados sin justificación)
* Proporciones reales en gráficos de área

#### 3. **Contexto**
* Títulos descriptivos
* Etiquetas de ejes claras
* Leyendas cuando sean necesarias
* Fuente y fecha de los datos

#### 4. **Jerarquía Visual**
* Lo más importante debe destacar
* Usa color estratégicamente para dirigir atención
* Contraste entre información principal y secundaria

## 2️⃣ Tipos de Gráficos y Sus Usos

### Gráficos de Comparación

#### **Gráfico de Barras** 📊
* **Uso**: Comparar cantidades entre categorías
* **Ejemplo**: Ventas por región, productos más vendidos
* **Variantes**: Horizontales (muchas categorías), agrupadas (múltiples series), apiladas (composición)

#### **Gráfico de Columnas** 📊
* **Uso**: Similar a barras, pero verticales
* **Ejemplo**: Ventas mensuales, evolución trimestral
* **Mejor para**: Pocas categorías (<10)

### Gráficos de Tendencia

#### **Gráfico de Líneas** 📈
* **Uso**: Mostrar evolución temporal
* **Ejemplo**: Ingresos a lo largo del tiempo, tendencias de KPIs
* **Ideal para**: Series temporales continuas

#### **Gráfico de Área** 📈
* **Uso**: Tendencia + magnitud
* **Ejemplo**: Acumulación de ventas, composición temporal
* **Variantes**: Simple, apilada (100% para proporciones)

### Gráficos de Distribución

#### **Histograma** 📊
* **Uso**: Distribución de una variable continua
* **Ejemplo**: Distribución de edades, precios
* **Muestra**: Frecuencia, asimetría, outliers

#### **Box Plot (Diagrama de Caja)** 📦
* **Uso**: Resumen estadístico de distribución
* **Ejemplo**: Comparar distribuciones entre grupos
* **Muestra**: Mediana, cuartiles, outliers

### Gráficos de Relación

#### **Scatter Plot (Dispersión)** 🔵
* **Uso**: Relación entre dos variables numéricas
* **Ejemplo**: Precio vs. demanda, edad vs. ingreso
* **Identifica**: Correlaciones, clusters, outliers

#### **Heatmap (Mapa de Calor)** 🔥
* **Uso**: Matrices de correlación, patrones bidimensionales
* **Ejemplo**: Correlación entre variables, actividad por hora/día

### Gráficos de Composición

#### **Gráfico de Torta (Pie Chart)** 🥧
* **Uso**: Proporciones de un total (¡úsalo con moderación!)
* **Ejemplo**: Cuota de mercado, distribución presupuestaria
* **Limitación**: Máximo 5-6 categorías

#### **Treemap** 🗺️
* **Uso**: Jerarquías y proporciones
* **Ejemplo**: Estructura de costos, categorización de productos
* **Ventaja**: Muchas categorías sin perder claridad

## 3️⃣ Matplotlib: Visualizaciones Personalizables

### ¿Qué es Matplotlib?

**Matplotlib** es la librería fundamental de visualización en Python:
* Base para muchas otras librerías (seaborn, pandas plotting)
* Control total sobre cada elemento del gráfico
* Sintaxis similar a MATLAB

### Anatomía de un Gráfico Matplotlib

```python
import matplotlib.pyplot as plt

# Crear figura y ejes
fig, ax = plt.subplots(figsize=(10, 6))

# Plotear datos
ax.plot(x, y, label='Serie 1')

# Personalizar
ax.set_title('Título del Gráfico')
ax.set_xlabel('Eje X')
ax.set_ylabel('Eje Y')
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```

### Dos Interfaces: pyplot vs OO

#### **pyplot (estilo MATLAB)** - Simple
```python
plt.plot(x, y)
plt.title('Título')
plt.show()
```

#### **Orientada a Objetos** - Recomendada
```python
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title('Título')
```

**💡 Mejor práctica**: Usa OO para gráficos complejos y múltiples subplots.

### Personalización Avanzada

* **Colores**: Paletas personalizadas, gradientes
* **Estilos**: `plt.style.use('seaborn')`, 'ggplot', 'fivethirtyeight'
* **Anotaciones**: Flechas, texto, marcadores especiales
* **Subplots**: Múltiples gráficos en una figura

## 4️⃣ Seaborn: Visualizaciones Estadísticas

### ¿Qué es Seaborn?

**Seaborn** es una librería de alto nivel construida sobre matplotlib:
* Visualizaciones estadísticas sofisticadas con menos código
* Estilos estéticos por defecto
* Integración nativa con pandas DataFrames

### Ventajas de Seaborn

1. **Sintaxis más simple** para gráficos complejos
2. **Temas visuales** atractivos out-of-the-box
3. **Análisis estadístico** automático (bandas de confianza, regresiones)
4. **Paletas de colores** diseñadas profesionalmente

## 5️⃣ Dashboards en Databricks

### ¿Qué es un Dashboard?

Un **dashboard** es una interfaz visual que:
* Consolida métricas clave (KPIs)
* Permite monitoreo en tiempo real
* Facilita decisiones rápidas
* Puede ser interactivo (filtros, drill-down)

### Databricks SQL Dashboards

**Características:**
* Basados en queries SQL
* Actualización automática
* Interactividad con parámetros
* Compartibles con stakeholders

### Componentes de un Buen Dashboard

#### 1. **Métricas Principales (KPIs)**
* Números grandes y visibles
* Comparación vs. periodo anterior
* Indicadores de tendencia (↑↓)

#### 2. **Gráficos de Contexto**
* Tendencias temporales
* Comparaciones entre categorías
* Distribuciones relevantes

#### 3. **Filtros Interactivos**
* Fecha/rango temporal
* Segmentación (región, producto, etc.)
* Múltiples dimensiones

#### 4. **Jerarquía de Información**
```
┌─────────────────────────────────────┐
│  KPIs Principales (grande)          │
├─────────────────────────────────────┤
│  Gráficos de Tendencia (mediano)    │
├─────────────────────────────────────┤
│  Detalles y Breakdown (pequeño)     │
└─────────────────────────────────────┘
```

### Mejores Prácticas de Dashboards

✅ **Un dashboard, una historia**
✅ **Máximo 5-7 visualizaciones** (evita sobrecarga)
✅ **Actualización clara** (timestamp, frecuencia)
✅ **Responsive design** (funciona en desktop y móvil)
❌ Evita colores estridentes
❌ No mezcles demasiadas métricas no relacionadas

## 6️⃣ Mejores Prácticas de Diseño Visual

### Color: Estrategia, No Decoración

#### **Uso Estratégico del Color**
* 🎯 **Resaltar insights**: Color brillante para lo importante
* 📊 **Categorías distintas**: Colores diferenciados
* 🌡️ **Escalas continuas**: Gradientes (frío-caliente)
* ⚠️ **Alertas**: Rojo para negativos/alertas, verde para positivos

#### **Paletas de Color**
* **Cualitativas**: Para categorías sin orden (nominal)
* **Secuenciales**: Para valores ordenados (bajo → alto)
* **Divergentes**: Para datos con punto medio (negativo ↔ positivo)

#### **Accesibilidad**
* ♿ Considera daltonismo (~8% hombres, ~0.5% mujeres)
* Usa herramientas: [ColorBrewer](https://colorbrewer2.org/), Viridis
* Prueba en escala de grises

### Tipografía y Etiquetas

* **Tamaños jerárquicos**: Título > Subtítulo > Etiquetas
* **Legibilidad**: Sans-serif para pantallas (Arial, Helvetica)
* **Contraste**: Texto oscuro sobre fondo claro (o viceversa)
* **No más de 2-3 fuentes** en un dashboard

### Espacio en Blanco (Whitespace)

* **Respira**: No llenes cada píxel
* **Agrupa**: Relacionados juntos, no relacionados separados
* **Márgenes**: Consistentes y proporcionales

### Alineación y Grid

* Alinea elementos en una grilla invisible
* Consistencia en tamaños y espaciados
* Simetría o asimetría intencional, no accidental

### Ratio de Data-Ink

**Principio de Edward Tufte:**
```
Data-Ink Ratio = Data ink / Total ink
```

* Maximiza la tinta dedicada a datos
* Minimiza elementos no informativos
* Elimina: Bordes innecesarios, fondos, efectos 3D, gridlines excesivas

## 7️⃣ Data Storytelling: Contar Historias con Datos

### ¿Qué es Data Storytelling?

**Data storytelling** combina:
1. 📊 **Datos**: Evidencia cuantitativa
2. 📈 **Visualización**: Representación visual clara
3. 📖 **Narrativa**: Historia que conecta emocionalmente

### Estructura de una Historia con Datos

#### 1. **Contexto** (Planteamiento)
* ¿Cuál es la situación?
* ¿Por qué es importante?
* ¿Quién es la audiencia?

#### 2. **Conflicto** (Nudo)
* ¿Qué problema existe?
* ¿Qué cambió o qué está en riesgo?
* Muestra datos que evidencian el problema

#### 3. **Resolución** (Desenlace)
* ¿Qué insights descubriste?
* ¿Qué acciones se recomiendan?
* ¿Cuál es el impacto esperado?

### Técnicas de Storytelling

#### **1. Orden Importancia → Detalle**
```
Insight Principal (grande)
     ↓
Evidencia (mediano)
     ↓
Detalles de apoyo (pequeño)
```

#### **2. Usa Comparaciones Relevantes**
* "Las ventas crecieron 25%" → "Las ventas crecieron 25%, el doble del crecimiento del mercado"
* Benchmarks, periodos anteriores, competencia

#### **3. Humaniza los Números**
* "15,000 clientes" → "15,000 familias confían en nosotros"
* "$2M en pérdidas" → "Suficiente para 50 nuevas contrataciones"

#### **4. Guía la Atención**
* Usa anotaciones (flechas, círculos, texto)
* Destaca con color el dato más importante
* Títulos descriptivos: "Las ventas cayeron 20% en Q3" > "Ventas Q3"

### Adaptación a la Audiencia

| Audiencia | Enfoque | Nivel de Detalle |
|-----------|---------|------------------|
| **Ejecutivos** | Insights, ROI, decisiones | Alto nivel |
| **Managers** | KPIs, tendencias, acciones | Medio |
| **Técnicos** | Metodología, datos crudos | Detallado |
| **General** | Historias, analogías | Simple |

### Checklist de Data Storytelling

✅ ¿El título comunica el insight principal?  
✅ ¿El gráfico es autoexplicativo?  
✅ ¿Hay un call-to-action claro?  
✅ ¿Usé la visualización más apropiada?  
✅ ¿Eliminé elementos distractores?  
✅ ¿El mensaje es memorable?
