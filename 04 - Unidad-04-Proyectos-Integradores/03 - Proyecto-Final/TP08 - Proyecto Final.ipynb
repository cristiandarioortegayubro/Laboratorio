````python
print("=" * 90)
print(" " * 20 + "PROYECTO FINAL: OPTIMIZACIÓN GEOESPACIAL")
print("=" * 90)

import pandas as pd
import numpy as np
import h3
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Cargar datos
ruta_datos = '/Workspace/Users/cortega@uda.edu.ar/Laboratorio/Datasets/'

df_clientes = pd.read_csv(ruta_datos + 'clientes.csv')
df_sucursales = pd.read_csv(ruta_datos + 'sucursales.csv')
df_ventas = pd.read_csv(ruta_datos + 'ventas.csv')

print(f"\n📊 Datasets cargados")
print(f"   Clientes: {len(df_clientes):,}")
print(f"   Sucursales actuales: {len(df_sucursales)}")
print(f"   Ventas: {len(df_ventas):,}")

print("\n" + "=" * 90)
print("FASE 1: ANÁLISIS DE COBERTURA ACTUAL")
print("=" * 90)

# Calcular cobertura actual
sucursales_h3 = df_sucursales['h3_index'].tolist()

def dist_sucursal_min(cliente_h3):
    if pd.isna(cliente_h3):
        return None
    try:
        return min([h3.grid_distance(cliente_h3, s) for s in sucursales_h3])
    except:
        return None

df_clientes['dist_sucursal'] = df_clientes['h3_index'].apply(dist_sucursal_min)

# Clasificar cobertura (distancia en hexágonos, res 9 ≈ 100m cada uno)
def clasificar_cobertura(dist):
    if pd.isna(dist):
        return 'Sin Datos'
    elif dist <= 5:  # ~500m
        return '✅ Excelente'
    elif dist <= 10:  # ~1km
        return '🟢 Buena'
    elif dist <= 20:  # ~2km
        return '🟡 Regular'
    else:
        return '🔴 Mala'

df_clientes['cobertura'] = df_clientes['dist_sucursal'].apply(clasificar_cobertura)

print(f"\n📋 Cobertura Actual:")
print(df_clientes['cobertura'].value_counts())

cobertura_buena = df_clientes['cobertura'].isin(['✅ Excelente', '🟢 Buena']).sum()
pct_cobertura = (cobertura_buena / len(df_clientes)) * 100

print(f"\n🎯 Métrica Clave:")
print(f"   Clientes con buena cobertura (<1km): {cobertura_buena} ({pct_cobertura:.1f}%)")
print(f"   🚨 Objetivo: 85% con buena cobertura")
print(f"   🔺 Gap: {85 - pct_cobertura:.1f}% puntos porcentuales")
````

````python
print("\n" + "=" * 90)
print("FASE 2: IDENTIFICAR ZONAS SIN COBERTURA ADECUADA")
print("=" * 90)

# Zonas con mala cobertura
zonas_sin_cobertura = df_clientes[
    df_clientes['cobertura'].isin(['🟡 Regular', '🔴 Mala'])
].copy()

print(f"\n📍 Clientes sin cobertura adecuada: {len(zonas_sin_cobertura):,}")

# Agregar por zona H3
zonas_prioritarias = zonas_sin_cobertura.groupby('h3_index').agg({
    'cliente_id': 'count',
    'dist_sucursal': 'mean',
    'latitud': 'first',
    'longitud': 'first'
}).reset_index()

zonas_prioritarias.columns = ['h3_index', 'num_clientes', 'dist_promedio', 'lat', 'lon']

# Agregar facturación potencial
ventas_geo = df_ventas[df_ventas['cliente_id'].notna()].merge(
    df_clientes[['cliente_id', 'h3_index']], 
    on='cliente_id'
)

facturacion = ventas_geo.groupby('h3_index')['total'].sum().reset_index()
facturacion.columns = ['h3_index', 'facturacion_historica']

zonas_prioritarias = zonas_prioritarias.merge(facturacion, on='h3_index', how='left')
zonas_prioritarias['facturacion_historica'] = zonas_prioritarias['facturacion_historica'].fillna(0)

# Score de prioridad: (clientes * facturación) / distancia
zonas_prioritarias['score_prioridad'] = (
    (zonas_prioritarias['num_clientes'] * zonas_prioritarias['facturacion_historica']) / 
    (zonas_prioritarias['dist_promedio'] + 1)  # +1 para evitar división por 0
)

zonas_prioritarias = zonas_prioritarias.sort_values('score_prioridad', ascending=False)

print(f"\n🎯 Top 10 Zonas Candidatas para Nueva Sucursal:")
print(zonas_prioritarias.head(10)[[
    'h3_index', 'num_clientes', 'facturacion_historica', 'dist_promedio', 'score_prioridad'
]])
````

````python
print("\n" + "=" * 90)
print("FASE 3: SELECCIÓN DE 2 UBICACIONES ÓPTIMAS")
print("=" * 90)

print(f"\n🧠 Algoritmo: Selección Greedy con Restricción de Distancia Mínima")
print(f"   1. Seleccionar zona con mayor score")
print(f"   2. Eliminar zonas muy cercanas (canibalización)")
print(f"   3. Repetir hasta tener 2 ubicaciones\n")

ubicaciones_seleccionadas = []
sucursales_totales = sucursales_h3.copy()
zonas_disponibles = zonas_prioritarias.copy()

DIST_MIN_ENTRE_SUCURSALES = 15  # ~1.5km mínimo entre sucursales

while len(ubicaciones_seleccionadas) < 2 and len(zonas_disponibles) > 0:
    # Seleccionar mejor zona
    mejor_zona = zonas_disponibles.iloc[0]
    ubicaciones_seleccionadas.append(mejor_zona)
    sucursales_totales.append(mejor_zona['h3_index'])
    
    print(f"🏪 Sucursal #{len(ubicaciones_seleccionadas)} seleccionada:")
    print(f"   H3: {mejor_zona['h3_index']}")
    print(f"   Ubicación: ({mejor_zona['lat']:.4f}, {mejor_zona['lon']:.4f})")
    print(f"   Clientes potenciales: {mejor_zona['num_clientes']}")
    print(f"   Facturación histórica: ${mejor_zona['facturacion_historica']:,.0f}")
    print()
    
    # Eliminar zona seleccionada
    zonas_disponibles = zonas_disponibles[zonas_disponibles['h3_index'] != mejor_zona['h3_index']]
    
    # Eliminar zonas muy cercanas
    def esta_muy_cerca(zona_h3):
        try:
            return h3.grid_distance(zona_h3, mejor_zona['h3_index']) < DIST_MIN_ENTRE_SUCURSALES
        except:
            return False
    
    zonas_disponibles = zonas_disponibles[
        ~zonas_disponibles['h3_index'].apply(esta_muy_cerca)
    ]

print(f"✅ 2 ubicaciones óptimas seleccionadas")
````

````python
print("\n" + "=" * 90)
print("FASE 4: PROYECCIÓN DE IMPACTO")
print("=" * 90)

# Recalcular cobertura con nuevas sucursales
def dist_sucursal_nueva(cliente_h3):
    if pd.isna(cliente_h3):
        return None
    try:
        return min([h3.grid_distance(cliente_h3, s) for s in sucursales_totales])
    except:
        return None

df_clientes['dist_sucursal_nueva'] = df_clientes['h3_index'].apply(dist_sucursal_nueva)
df_clientes['cobertura_nueva'] = df_clientes['dist_sucursal_nueva'].apply(clasificar_cobertura)

print(f"\n📋 Comparación: Antes vs. Después\n")

cobertura_comparacion = pd.DataFrame({
    'Antes': df_clientes['cobertura'].value_counts(),
    'Después': df_clientes['cobertura_nueva'].value_counts()
})

print(cobertura_comparacion)

# Métrica principal
cobertura_buena_nueva = df_clientes['cobertura_nueva'].isin(['✅ Excelente', '🟢 Buena']).sum()
pct_cobertura_nueva = (cobertura_buena_nueva / len(df_clientes)) * 100

print(f"\n🎯 Métrica de Éxito:")
print(f"   Antes:   {cobertura_buena} clientes ({pct_cobertura:.1f}%)")
print(f"   Después: {cobertura_buena_nueva} clientes ({pct_cobertura_nueva:.1f}%)")
print(f"   📈 Mejora: +{cobertura_buena_nueva - cobertura_buena} clientes (+{pct_cobertura_nueva - pct_cobertura:.1f}pp)")

if pct_cobertura_nueva >= 85:
    print(f"   ✅ OBJETIVO ALCANZADO: {pct_cobertura_nueva:.1f}% ≥ 85%")
else:
    print(f"   ⚠️ Objetivo no alcanzado: {pct_cobertura_nueva:.1f}% < 85%")

# Estimación de facturación incremental
clientes_mejorados = df_clientes[
    (df_clientes['cobertura'].isin(['🟡 Regular', '🔴 Mala'])) &
    (df_clientes['cobertura_nueva'].isin(['✅ Excelente', '🟢 Buena']))
]

print(f"\n💰 Proyección de Facturación:")
print(f"   Clientes con cobertura mejorada: {len(clientes_mejorados)}")

# Asumiendo 15% aumento en frecuencia de compra
ticket_promedio = df_ventas['total'].mean()
frecuencia_actual = len(df_ventas) / len(df_clientes) / 2  # ventas por cliente por año
facturacion_incremental = len(clientes_mejorados) * ticket_promedio * frecuencia_actual * 0.15 * 2

print(f"   Facturación incremental anual estimada: ${facturacion_incremental:,.0f}")
````

````python
print("\n" + "=" * 90)
print("FASE 5: VISUALIZACIÓN GEOESPACIAL")
print("=" * 90)

print(f"\n🗺️ Creando mapa interactivo con recomendaciones...\n")

# Crear mapa base
fig = go.Figure()

# 1. Sucursales actuales
fig.add_trace(go.Scattermapbox(
    lat=df_sucursales['latitud'],
    lon=df_sucursales['longitud'],
    mode='markers+text',
    marker=dict(size=20, color='blue', symbol='star'),
    text=df_sucursales['nombre'].apply(lambda x: x.split('-')[1].strip()),
    textposition='top center',
    name='Sucursales Actuales',
    hovertext=df_sucursales['nombre'],
    hoverinfo='text'
))

# 2. Nuevas sucursales recomendadas
df_nuevas = pd.DataFrame(ubicaciones_seleccionadas)

fig.add_trace(go.Scattermapbox(
    lat=df_nuevas['lat'],
    lon=df_nuevas['lon'],
    mode='markers+text',
    marker=dict(size=25, color='red', symbol='star', line=dict(width=2, color='white')),
    text=['NUEVA 1', 'NUEVA 2'],
    textposition='bottom center',
    name='Sucursales Recomendadas',
    hovertext=df_nuevas.apply(
        lambda row: f"Nueva Sucursal<br>Clientes: {row['num_clientes']}<br>Facturación: ${row['facturacion_historica']:,.0f}",
        axis=1
    ),
    hoverinfo='text'
))

# 3. Zonas sin cobertura (top 30)
top_zonas_viz = zonas_prioritarias.head(30)

fig.add_trace(go.Scattermapbox(
    lat=top_zonas_viz['lat'],
    lon=top_zonas_viz['lon'],
    mode='markers',
    marker=dict(
        size=top_zonas_viz['num_clientes']*2,
        color=top_zonas_viz['score_prioridad'],
        colorscale='YlOrRd',
        opacity=0.6,
        colorbar=dict(title='Score Prioridad')
    ),
    name='Zonas sin Cobertura',
    hovertext=top_zonas_viz.apply(
        lambda row: f"Zona sin cobertura<br>Clientes: {row['num_clientes']}<br>Dist actual: {row['dist_promedio']:.1f} hex",
        axis=1
    ),
    hoverinfo='text'
))

fig.update_layout(
    mapbox=dict(
        style='open-street-map',
        center=dict(lat=-32.89, lon=-68.84),
        zoom=11
    ),
    title='Mapa Estratégico: Ubicaciones Óptimas para Nuevas Sucursales',
    height=700,
    showlegend=True
)

fig.show()

print("✅ Mapa interactivo generado")
````

````python
# CARGA Y PREPARACIÓN DE DATOS
print("\n📂 FASE 1: CARGA Y PREPARACIÓN DE DATOS")
print("=" * 80)

ruta_datos = '/Workspace/Users/cortega@uda.edu.ar/Laboratorio/Datasets/'

# Cargar datasets
df_productos = pd.read_csv(ruta_datos + 'productos.csv')
df_sucursales = pd.read_csv(ruta_datos + 'sucursales.csv')
df_clientes = pd.read_csv(ruta_datos + 'clientes.csv', parse_dates=['fecha_registro'])
df_ventas = pd.read_csv(ruta_datos + 'ventas.csv', parse_dates=['fecha'])
df_detalles = pd.read_csv(ruta_datos + 'detalles_ventas.csv')

# Consolidar dataset maestro
df_maestro = df_detalles.merge(df_ventas[['venta_id', 'fecha', 'sucursal_id', 'cliente_id']], on='venta_id')
df_maestro = df_maestro.merge(df_productos[['producto_id', 'nombre', 'categoria', 'costo_unitario']], on='producto_id')
df_maestro = df_maestro.merge(df_sucursales[['sucursal_id', 'zona']], on='sucursal_id')

# Calcular métricas de rentabilidad
df_maestro['costo_total'] = df_maestro['costo_unitario'] * df_maestro['cantidad']
df_maestro['ganancia'] = df_maestro['subtotal'] - df_maestro['costo_total']
df_maestro['margen_%'] = (df_maestro['ganancia'] / df_maestro['subtotal'] * 100).round(2)

print(f"\n✅ Datos consolidados: {len(df_maestro):,} registros")
print(f"\n📊 Resumen general:")
print(f"  Facturación total: ${df_maestro['subtotal'].sum():,.2f}")
print(f"  Ganancia total: ${df_maestro['ganancia'].sum():,.2f}")
print(f"  Margen promedio: {(df_maestro['ganancia'].sum() / df_maestro['subtotal'].sum() * 100):.2f}%")
````

## 📉 ANÁLISIS 1: Matriz de Rentabilidad por Producto

### Objetivo
Clasificar productos según su contribución a ventas y rentabilidad

### Criterios de Clasificación

* **ESTRELLA** ⭐: Alto volumen + Alto margen (mantener y promover)
* **VACA LECHERA** 🐄: Alto volumen + Bajo margen (optimizar precio/costo)
* **INTERROGANTE** ❓: Bajo volumen + Alto margen (aumentar visibilidad)
* **PERRO** 🐶: Bajo volumen + Bajo margen (descontinuar o replantear)

````python
# ANÁLISIS DE RENTABILIDAD POR PRODUCTO
print("\n📊 FASE 2: ANÁLISIS DE RENTABILIDAD POR PRODUCTO")
print("=" * 80)

# Agregar por producto
analisis_productos = df_maestro.groupby(['producto_id', 'nombre', 'categoria']).agg({
    'cantidad': 'sum',
    'subtotal': 'sum',
    'ganancia': 'sum',
    'venta_id': 'nunique'
}).reset_index()

analisis_productos.columns = ['producto_id', 'nombre', 'categoria', 'unidades_vendidas', 
                               'facturacion', 'ganancia', 'num_ventas']

analisis_productos['margen_%'] = (analisis_productos['ganancia'] / analisis_productos['facturacion'] * 100).round(2)
analisis_productos['contribucion_facturacion_%'] = (analisis_productos['facturacion'] / analisis_productos['facturacion'].sum() * 100).round(2)
analisis_productos['contribucion_ganancia_%'] = (analisis_productos['ganancia'] / analisis_productos['ganancia'].sum() * 100).round(2)

print(f"\n✅ Análisis de {len(analisis_productos)} productos completado")
print("\n🔝 Top 10 productos por facturación:")
display(analisis_productos.nlargest(10, 'facturacion')[['nombre', 'categoria', 'unidades_vendidas', 'facturacion', 'margen_%']])
````

````python
# Clasificar productos en cuadrantes
mediana_volumen = analisis_productos['facturacion'].median()
mediana_margen = analisis_productos['margen_%'].median()

def clasificar_producto(row):
    if row['facturacion'] >= mediana_volumen and row['margen_%'] >= mediana_margen:
        return '⭐ ESTRELLA'
    elif row['facturacion'] >= mediana_volumen and row['margen_%'] < mediana_margen:
        return '🐄 VACA LECHERA'
    elif row['facturacion'] < mediana_volumen and row['margen_%'] >= mediana_margen:
        return '❓ INTERROGANTE'
    else:
        return '🐶 PERRO'

analisis_productos['clasificacion'] = analisis_productos.apply(clasificar_producto, axis=1)

print("\n🎯 CLASIFICACIÓN DE PRODUCTOS")
print("=" * 80)
print("\n📊 Distribución por cuadrante:")
print(analisis_productos['clasificacion'].value_counts())

print("\n🐶 PRODUCTOS 'PERRO' (candidatos a descontinuar):")
perros = analisis_productos[analisis_productos['clasificacion'] == '🐶 PERRO'].sort_values('facturacion')
display(perros[['nombre', 'categoria', 'facturacion', 'margen_%', 'unidades_vendidas']].head(10))
````

````python
# Visualizar matriz de portafolio
fig, ax = plt.subplots(figsize=(14, 8))

# Colores por clasificación
color_map = {
    '⭐ ESTRELLA': '#FFD700',
    '🐄 VACA LECHERA': '#4CAF50',
    '❓ INTERROGANTE': '#2196F3',
    '🐶 PERRO': '#F44336'
}

for clasificacion, color in color_map.items():
    subset = analisis_productos[analisis_productos['clasificacion'] == clasificacion]
    ax.scatter(subset['facturacion'], subset['margen_%'], 
               s=subset['unidades_vendidas']/10, alpha=0.6, 
               c=color, label=clasificacion, edgecolors='black', linewidth=0.5)

# Líneas de medianas
ax.axvline(mediana_volumen, color='gray', linestyle='--', linewidth=1, alpha=0.7, label='Mediana Facturación')
ax.axhline(mediana_margen, color='gray', linestyle='--', linewidth=1, alpha=0.7, label='Mediana Margen')

ax.set_xlabel('Facturación Total ($)', fontsize=12, fontweight='bold')
ax.set_ylabel('Margen (%)', fontsize=12, fontweight='bold')
ax.set_title('🎯 Matriz de Portafolio de Productos\n(Tamaño = Unidades Vendidas)', 
             fontsize=14, fontweight='bold', pad=20)
ax.legend(loc='best', fontsize=10)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✅ Matriz de portafolio generada")
````

## 👥 ANÁLISIS 2: Segmentación de Clientes (RFM)

### Objetivo
Segmentar clientes según comportamiento de compra para estrategias diferenciadas

### Método RFM

* **R (Recency)**: Días desde la última compra
* **F (Frequency)**: Número total de compras
* **M (Monetary)**: Gasto total acumulado

### Segmentos Esperados

* **Campeones** 🏆: R alto, F alto, M alto (clientes más valiosos)
* **Leales** 💖: F alto (compran frecuentemente)
* **Nuevos** 🌱: R alto, F bajo (recién llegados)
* **En Riesgo** ⚠️: R bajo (no compran hace tiempo)
* **Perdidos** ❌: R muy bajo, F bajo

````python
# SEGMENTACIÓN DE CLIENTES RFM
print("\n👥 FASE 3: SEGMENTACIÓN DE CLIENTES (RFM)")
print("=" * 80)

# Filtrar solo ventas con cliente identificado
df_clientes_ventas = df_maestro[df_maestro['cliente_id'].notna()].copy()

# Fecha de referencia (fecha más reciente + 1 día)
fecha_referencia = df_clientes_ventas['fecha'].max() + pd.Timedelta(days=1)

# Calcular RFM por cliente
rfm = df_clientes_ventas.groupby('cliente_id').agg({
    'fecha': lambda x: (fecha_referencia - x.max()).days,  # Recency
    'venta_id': 'nunique',  # Frequency
    'subtotal': 'sum'  # Monetary
}).reset_index()

rfm.columns = ['cliente_id', 'recency', 'frequency', 'monetary']

# Join con info de clientes
rfm = rfm.merge(df_clientes[['cliente_id', 'es_vip']], on='cliente_id', how='left')

print(f"\n✅ Análisis RFM de {len(rfm)} clientes completado")
print("\n📊 Estadísticas RFM:")
print(rfm[['recency', 'frequency', 'monetary']].describe().round(2))
````

````python
# Crear scores RFM (1-5, donde 5 es mejor)
rfm['R_score'] = pd.qcut(rfm['recency'], 5, labels=[5, 4, 3, 2, 1])  # Invertido: menor recency = mejor
rfm['F_score'] = pd.qcut(rfm['frequency'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5])
rfm['M_score'] = pd.qcut(rfm['monetary'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5])

# Convertir a int
rfm['R_score'] = rfm['R_score'].astype(int)
rfm['F_score'] = rfm['F_score'].astype(int)
rfm['M_score'] = rfm['M_score'].astype(int)

# Score RFM combinado
rfm['RFM_score'] = rfm['R_score'].astype(str) + rfm['F_score'].astype(str) + rfm['M_score'].astype(str)
rfm['RFM_score_num'] = rfm['R_score'] + rfm['F_score'] + rfm['M_score']

# Segmentar clientes
def segmentar_cliente(row):
    if row['RFM_score_num'] >= 13:
        return '🏆 Campeones'
    elif row['F_score'] >= 4:
        return '💖 Leales'
    elif row['R_score'] >= 4 and row['F_score'] <= 2:
        return '🌱 Nuevos'
    elif row['R_score'] <= 2 and row['F_score'] >= 3:
        return '⚠️ En Riesgo'
    elif row['R_score'] <= 2:
        return '❌ Perdidos'
    else:
        return '👤 Regulares'

rfm['segmento'] = rfm.apply(segmentar_cliente, axis=1)

print("\n🎯 SEGMENTACIÓN DE CLIENTES")
print("=" * 80)
print("\n📊 Distribución por segmento:")
segmentacion = rfm.groupby('segmento').agg({
    'cliente_id': 'count',
    'monetary': 'sum',
    'frequency': 'mean'
}).round(2)
segmentacion.columns = ['Num_Clientes', 'Valor_Total', 'Frecuencia_Promedio']
segmentacion['Valor_%'] = (segmentacion['Valor_Total'] / segmentacion['Valor_Total'].sum() * 100).round(2)
display(segmentacion.sort_values('Valor_Total', ascending=False))
````

````python
# Visualizar segmentos
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Gráfico 1: Número de clientes por segmento
segmento_counts = rfm['segmento'].value_counts()
colors = ['#FFD700', '#4CAF50', '#8BC34A', '#FFC107', '#FF5722', '#9E9E9E']
ax1.bar(range(len(segmento_counts)), segmento_counts.values, color=colors, edgecolor='black')
ax1.set_xticks(range(len(segmento_counts)))
ax1.set_xticklabels(segmento_counts.index, rotation=45, ha='right')
ax1.set_ylabel('Número de Clientes', fontsize=12, fontweight='bold')
ax1.set_title('👥 Distribución de Clientes por Segmento', fontsize=14, fontweight='bold')
ax1.grid(axis='y', alpha=0.3)
for i, v in enumerate(segmento_counts.values):
    ax1.text(i, v, str(v), ha='center', va='bottom', fontsize=10, fontweight='bold')

# Gráfico 2: Valor total por segmento
segmento_valor = rfm.groupby('segmento')['monetary'].sum().sort_values(ascending=False)
ax2.barh(range(len(segmento_valor)), segmento_valor.values, color=colors, edgecolor='black')
ax2.set_yticks(range(len(segmento_valor)))
ax2.set_yticklabels(segmento_valor.index)
ax2.set_xlabel('Valor Total ($)', fontsize=12, fontweight='bold')
ax2.set_title('💰 Valor Total Generado por Segmento', fontsize=14, fontweight='bold')
ax2.grid(axis='x', alpha=0.3)
for i, v in enumerate(segmento_valor.values):
    ax2.text(v, i, f'${v:,.0f}', ha='left', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()

print("\n✅ Segmentación visual completada")
````
