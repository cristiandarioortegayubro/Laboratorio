### 📊 Módulo 02: Monitoreo de Modelos en Producción

**🎯 Objetivos:**
* Entender degradación de modelos
* Detectar data drift y concept drift
* Monitorear performance en tiempo real
* Implementar alertas automáticas

**📓 Notebooks:**

#### [02-Teoria-Monitoreo-Modelos-Produccion](#notebook-1402451318137608)
* **Contenido**:
  * **¿Por qué Monitorear?**:
    * Degradación de modelos en producción
    * Causas: data drift, concept drift, errores upstream
  * **Data Drift**:
    * Covariate shift (P(X) cambia)
    * Prior probability shift (P(Y) cambia)
    * Detección: KS test, PSI, Wasserstein distance
  * **Concept Drift**:
    * Relación X → Y cambia
    * Tipos: Sudden, Gradual, Incremental, Recurring
    * Detección: monitorear accuracy con ground truth
  * **Métricas de Monitoreo**:
    * Performance metrics (accuracy, precision, recall)
    * Proxy metrics (sin ground truth)
    * System metrics (latencia, throughput, memoria)
  * **Estrategias de Respuesta**:
    * Reentrenar modelo
    * Ajustar threshold
    * Feature recalibration
    * Ensemble con modelo nuevo
  * **Arquitectura de Monitoreo**:
    * Prediction logs → Drift detection → Alertas
  * **Mejores Prácticas**:
    * Log everything
    * Ventanas móviles
    * Múltiples detectores
    * Automatic retraining

#### [02-Practica-Monitoreo-Modelos-Produccion](#notebook-1402451318137609)
* **Ejercicios**:
  1. Detectar Data Drift con KS test
  2. Calcular PSI (Population Stability Index)
  3. Visualizar distribuciones (con/sin drift)
  4. Simulación de Concept Drift
  5. Dashboard de monitoreo completo
* **Datasets**: Datos sintéticos con drift simulado

**⏱️ Duración**: 90-120 minutos

### 🧪 Módulo 03: A/B Testing y Experimentación

**🎯 Objetivos:**
* Diseñar experimentos A/B válidos
* Calcular tamaño de muestra
* Analizar resultados con tests estadísticos
* Evitar pitfalls comunes

**📓 Notebooks:**

#### [03-Teoria-AB-Testing-Experimentacion](#notebook-1402451318137610)
* **Contenido**:
  * **¿Qué es A/B Testing?**:
    * Experimento controlado (A vs B)
    * Ejemplos en ML: modelos, features, hiperparámetros
  * **Diseño de Experimentos**:
    * Hipótesis (H0, H1)
    * Elegir métrica primaria
    * Calcular tamaño de muestra (power analysis)
    * Asignación aleatoria (hash determinista)
  * **Análisis Estadístico**:
    * Test de proporciones (Z-test)
    * Test t para métricas continuas
    * Intervalos de confianza
  * **Pitfalls Comunes**:
    * Peeking (mirar antes de tiempo)
    * Multiple testing (corrección Bonferroni)
    * Novelty effect
    * Selection bias
  * **A/B Testing para Modelos ML**:
    * Shadow mode
    * Canary deployment
    * Multi-Armed Bandit
  * **Checklist de A/B Testing** (10 puntos)

#### [03-Practica-AB-Testing-Experimentacion](#notebook-1402451318137611)
* **Ejercicios**:
  1. Simular experimento A/B
  2. Test de proporciones
  3. Calcular lift y significancia
* **Métricas**: Conversion rate, Z-test, p-value

**⏱️ Duración**: 75-90 minutos
