## 🗺️ Parte 3.5: Análisis Geoespacial con H3

### ¿Qué es H3?

H3 es un sistema de indexación geoespacial desarrollado por Uber que **divide la Tierra en hexágonos**. Cada ubicación GPS (latitud, longitud) se convierte en un **índice H3 único**.

**¿Por qué hexágonos y no cuadrados?**
- 🟢 Todos los vecinos están a la misma distancia del centro
- 🟢 Mejores para representar áreas de cobertura circulares
- 🟢 Más precisos para cálculos de distancia

**Resoluciones H3:**
- Resolución 0: ~4,000 km² por hexágono (escala continental)
- Resolución 5: ~25 km² (escala ciudad)
- **Resolución 9: ~0.1 km²** ← *Usamos esta para nuestra panadería*
- Resolución 15: ~1 m² (escala edificio)

### Datos Geoespaciales en Nuestros Datasets

Los datasets de **sucursales** y **clientes** ahora incluyen:
- `latitud`, `longitud`: Coordenadas GPS
- `h3_index`: Índice H3 (resolución 9)

```python
# Ver las columnas geoespaciales en sucursales
print("🏪 SUCURSALES - Datos Geoespaciales")
print("=" * 60)
print(df_sucursales[['nombre', 'zona', 'latitud', 'longitud', 'h3_index']])

# Mostrar coordenadas en formato legible
print("\n📍 Coordenadas GPS de las sucursales:")
for idx, row in df_sucursales.iterrows():
    print(f"  {row['nombre']}:")
    print(f"    GPS: ({row['latitud']}, {row['longitud']})")
    print(f"    H3:  {row['h3_index']}")
```

```python
# Calcular distancias entre sucursales usando H3
print("📏 DISTANCIAS ENTRE SUCURSALES (en número de hexágonos H3)")
print("=" * 60)

sucursales_list = df_sucursales[['nombre', 'h3_index']].to_dict('records')

for i, suc1 in enumerate(sucursales_list):
    for suc2 in sucursales_list[i+1:]:
        distancia = h3.grid_distance(suc1['h3_index'], suc2['h3_index'])
        print(f"\n{suc1['nombre']} ↔ {suc2['nombre']}:")
        print(f"  Distancia: {distancia} hexágonos H3 (~{distancia * 0.1:.1f} km aproximadamente)")

print("\n💡 TIP: Cada hexágono de resolución 9 tiene ~0.1 km²")
print("   La distancia real puede variar según la topografía y rutas")
```

```python
# Ver muestra de clientes con datos geoespaciales
print("👥 CLIENTES - Datos Geoespaciales (muestra de 10)")
print("=" * 80)
print(df_clientes[['cliente_id', 'nombre', 'latitud', 'longitud', 'h3_index']].head(10))

# Estadísticas de distribución geográfica
print("\n📊 DISTRIBUCIÓN GEOGRÁFICA DE CLIENTES")
print("=" * 60)
print(f"Total de clientes: {len(df_clientes):,}")
print(f"Hexágonos H3 únicos ocupados: {df_clientes['h3_index'].nunique():,}")
print(f"Promedio de clientes por hexágono: {len(df_clientes) / df_clientes['h3_index'].nunique():.2f}")

# Top 5 hexágonos con más clientes
print("\n🔝 Top 5 zonas con mayor concentración de clientes:")
top_h3 = df_clientes['h3_index'].value_counts().head(5)
for idx, (h3_index, count) in enumerate(top_h3.items(), 1):
    print(f"  {idx}. Hexágono {h3_index}: {count} clientes")
```

### ✅ Ejercicio Práctico: Área de Cobertura

**Objetivo:** Calcular cuántos clientes están en el área de cobertura de cada sucursal.

Usaremos `h3.grid_disk(h3_index, radio)` para obtener todos los hexágonos vecinos dentro de un radio.

```python
# Para cada sucursal, calcular clientes en área de cobertura (radio 2)
print("📍 ÁREA DE COBERTURA DE SUCURSALES (radio 2 hexágonos)")
print("=" * 70)

for idx, row in df_sucursales.iterrows():
    # Obtener hexágonos vecinos en radio 2
    vecinos = h3.grid_disk(row['h3_index'], 2)
    
    # Contar clientes en esa área
    clientes_cercanos = df_clientes[df_clientes['h3_index'].isin(vecinos)]
    
    print(f"\n{row['nombre']} ({row['zona']})")
    print(f"  🔶 Hexágonos en área de cobertura: {len(vecinos)}")
    print(f"  👥 Clientes en área: {len(clientes_cercanos)} ({len(clientes_cercanos)/len(df_clientes)*100:.1f}%)")

print("\n💡 INTERPRETACIÓN:")
print("   - Radio 2 hexágonos ≈ ~0.2-0.3 km de cobertura")
print("   - Útil para planificar zonas de delivery")
print("   - Identificar áreas sin cobertura adecuada")
```
